### **Problema 1**
Supongamos un haz de fotones de 1 MeV incidiendo sobre un disco de aluminio ($\rho$ = 2.70 g cm$^{-3}$) de 10 cm de espesor y 4 cm de diámetro. La sección eficaz total de interacción es $\sigma$=2.75$\times$10$^{-24}$cm$^{2}$ y el peso atómico Pa(Al)=27.0 g.

a) Si un solo fotón incide perpendicularmente sobre la lámina ¿Qué probabilidad tiene de interaccionar dentro del primer centímetro de recorrido? ¿Qué probabilidad tiene de interaccionar antes de escapar del blanco?.

b) Un haz cilíndrico de 1.2$\times$10$^4$ fotones cm$^{-2}$ y 2 cm de diámetro incide perpendicularmente en el centro del disco de aluminio ¿Cuantos fotones del haz interaccionan con el blanco? Resolver el mismo problema para un haz de 10 cm de diámetro (mayor que el blanco).




---


#### a) Probabilidad de interacción de un solo fotón

Para analizar la probabilidad de interacción de un único fotón en el disco de aluminio, primero calculamos el coeficiente de atenuación lineal ($\mu$). Utilizando el número de Avogadro proporcionado en los datos generales de la hoja ($N_a = 6.02 \times 10^{23}$), la densidad atómica es $n = \frac{N_a \cdot \rho}{P_a} = \frac{(6.02 \times 10^{23}) \cdot 2.70}{27.0} = 6.02 \times 10^{22} \text{ átomos/cm}^3$. Multiplicando por la sección eficaz total ($\sigma = 2.75 \times 10^{-24} \text{ cm}^2$), obtenemos un coeficiente **$\mu = 0.16555 \text{ cm}^{-1}$**.

Al evaluar las probabilidades solicitadas para este haz de 1 MeV, podemos comparar el uso de la aproximación matemática de **blanco fino** frente al cálculo exacto de **blanco grueso**:

### 1. Aproximación de blanco fino ($P \approx \mu \cdot x$)
Esta aproximación se deriva del desarrollo en serie de Taylor de la exponencial y asume que la probabilidad de interacción crece linealmente con el espesor. Matemáticamente solo es válida cuando $\mu x \ll 1$ (es decir, cuando el haz apenas sufre atenuación).

*   **Probabilidad en el primer centímetro ($x = 1 \text{ cm}$):**
    $P \approx 0.16555 \cdot 1 = \mathbf{0.1655 \text{ (16.55\%)}}$
*   **Probabilidad antes de escapar del blanco ($x = 10 \text{ cm}$):**
    $P \approx 0.16555 \cdot 10 = \mathbf{1.6555 \text{ (165.55\%)}}$

Como se puede observar, para 10 cm de espesor el resultado de esta aproximación pierde todo el sentido físico (arroja una probabilidad mayor al 100%). El material es demasiado masivo para sostener la aproximación lineal.

### 2. Cálculo exacto o de blanco grueso ($P = 1 - e^{-\mu x}$)
Este cálculo aplica rigurosamente la ley de atenuación exponencial, la cual contabiliza la reducción exponencial de los fotones del haz a medida que van penetrando en el medio. Es válido para cualquier espesor.

*   **Probabilidad en el primer centímetro ($x = 1 \text{ cm}$):**
    $P = 1 - e^{-0.16555\text{ cm}^{-1} \cdot 10 \text{ cm}} = \mathbf{0.1526 \text{ (15.26\%)}}$
*   **Probabilidad antes de escapar del blanco ($x = 10 \text{ cm}$):**
    $P = 1 - e^{-1.6555 \text{ cm}^{-1} \cdot 10 \text{ cm}} = 1 - 0.1910 = \mathbf{0.8090 \text{ (80.90\%)}}$.

### Conclusión de la comparación
Para el **primer centímetro**, la aproximación de blanco fino (16.55%) ofrece una estimación moderadamente cercana al valor real (15.26%), aunque lo sobrestima por no tener en cuenta a los fotones que ya han interaccionado en las capas más superficiales de ese primer centímetro.

Sin embargo, al evaluar todo el **blanco de 10 cm de espesor**, el régimen de "blanco fino" fracasa por completo. Para calcular interacciones en un espesor donde $\mu x$ no es despreciable (en este caso $\mu x = 1.6555$), **es estrictamente necesario aplicar el tratamiento de blanco grueso**, obteniendo así la verdadera probabilidad de interacción del **80.90%**.


```python
# Coeficiente de atenuación lineal calculado (Aluminio a 1 MeV)
mu = 0.16555  # cm^-1

# Rango de profundidades x en centímetros (de 0 a 10 cm)
x = np.linspace(0, 10, 100)

# 1. Aproximación de blanco fino (Probabilidad lineal)
# Nota: Carece de sentido físico cuando supera 1 (100%)
P_fino = mu * x

# 2. Cálculo exacto de blanco grueso (Probabilidad exponencial)
P_grueso = 1 - np.exp(-mu * x)

# Configuración y creación del gráfico
plt.figure(figsize=(9, 6))

# Dibujar las curvas
plt.plot(x, P_fino, label=r'Blanco Fino ($P \approx \mu \cdot x$)',
         linestyle='--', color='red', linewidth=2)
plt.plot(x, P_grueso, label=r'Blanco Grueso ($P = 1 - e^{-\mu \cdot x}$)',
         linestyle='-', color='blue', linewidth=2)

# Línea horizontal que marca el límite físico lógico (100% de probabilidad)
plt.axhline(1, color='gray', linestyle=':', label='Límite físico (Probabilidad = 1)')

# Etiquetas, título y leyenda
plt.title('Comparación de Probabilidad de Interacción: Blanco Fino vs. Grueso\n' + r'(Fotones de 1 MeV en Aluminio, $\mu = 0.16555$ cm$^{-1}$)')
plt.xlabel('Profundidad $x$ (cm)')
plt.ylabel('Probabilidad de Interacción $P(x)$')

# Limitar los ejes para una mejor visualización de la divergencia
plt.xlim(0, 10)
plt.ylim(0, 1.8)

plt.legend(loc='upper left')
plt.grid(True, alpha=0.5)

# Mostrar la gráfica
plt.tight_layout()
plt.show()
```

#### b) Fotones del haz que interaccionan con el blanco

Tenemos una fluencia incidente $\Phi = 1.2 \times 10^4 \text{ fotones/cm}^2$. El número de fotones que interaccionan dependerá de cuántos fotones *impacten efectivamente* en el blanco, multiplicado por la probabilidad de interacción calculada antes para los 10 cm de espesor ($P = 0.8090$).

**Caso 1: Haz de 2 cm de diámetro**

Como el diámetro del haz (2 cm) es **menor** que el diámetro del blanco (4 cm), todos los fotones del haz inciden en el disco de aluminio.
*   **Área transversal del haz:** $A_{haz} = \pi \cdot r^2 = \pi \cdot (1 \text{ cm})^2 = \pi \text{ cm}^2 \approx 3.1416 \text{ cm}^2$
*   **Fotones totales que inciden:** $N_{inc} = \Phi \cdot A_{haz} = (1.2 \times 10^4) \cdot \pi \approx 37,699 \text{ fotones}$
*   **Fotones que interaccionan:** $N_{int} = N_{inc} \cdot P(10\text{ cm}) = 37,699 \cdot 0.8090 \approx \mathbf{30,499 \text{ fotones}}$

**Caso 2: Haz de 10 cm de diámetro**

En este caso, el haz (10 cm) es **mayor** que el disco de aluminio (4 cm). Los fotones que caen por fuera del disco no pueden interaccionar. Por lo tanto, el área efectiva de incidencia viene limitada por el propio blanco.
*   **Área efectiva de incidencia (el blanco):** $A_{blanco} = \pi \cdot r_{blanco}^2 = \pi \cdot (2 \text{ cm})^2 = 4\pi \text{ cm}^2 \approx 12.566 \text{ cm}^2$
*   **Fotones que logran golpear el blanco:** $N_{inc} = \Phi \cdot A_{blanco} = (1.2 \times 10^4) \cdot 4\pi \approx 150,796 \text{ fotones}$
*   **Fotones que interaccionan:** $N_{int} = N_{inc} \cdot P(10\text{ cm}) = 150,796 \cdot 0.8090 \approx \mathbf{121,995 \text{ fotones}}$

```python
# Constantes del problema 1 (Aluminio a 1 MeV)
mu = 0.16555  # cm^-1 (coeficiente de atenuación)
phi = 1.2e4   # fotones/cm^2 (fluencia del haz)

def draw_cylinder_3d(ax, radius, z_min, z_max, color, alpha, label=''):
    """Función auxiliar para generar un cilindro en 3D"""
    z = np.linspace(z_min, z_max, 50)
    theta = np.linspace(0, 2*np.pi, 50)
    theta_grid, z_grid = np.meshgrid(theta, z)
    x_grid = radius * np.cos(theta_grid)
    y_grid = radius * np.sin(theta_grid)
    ax.plot_surface(x_grid, y_grid, z_grid, color=color, alpha=alpha, edgecolor='none', label=label)

def calcular_interacciones_3d(d_haz, d_blanco, x_espesor):
    # 1. Cálculos físicos exactos (blanco grueso)
    P = 1 - np.exp(-mu * x_espesor)

    # Lógica del diámetro efectivo
    d_efectivo = min(d_haz, d_blanco)
    area_efectiva = np.pi * (d_efectivo / 2)**2
    n_inc = phi * area_efectiva
    n_int = P * n_inc

    # 2. Mostrar resultados numéricos por pantalla
    print("-" * 60)
    print(f"Probabilidad de interacción (espesor {x_espesor} cm): {P*100:.2f}%")

    if d_haz > d_blanco:
        print(f"\n¡Atención! El haz ({d_haz} cm) desborda el blanco ({d_blanco} cm).")
        print(f"El volumen de interacción queda limitado por el blanco.")
    else:
        print(f"\nEl haz penetra completamente dentro del blanco.")

    print(f"\nFotones incidentes sobre el blanco: {n_inc:,.0f}")
    print(f"Fotones que interaccionan:          {n_int:,.0f}")
    print("-" * 60)

    # 3. Representación Gráfica 3D
    fig = plt.figure(figsize=(10, 8))
    ax = fig.add_subplot(111, projection='3d')

    # A) Dibujar el blanco (Aluminio) - Cilindro Gris
    draw_cylinder_3d(ax, radius=d_blanco/2, z_min=0, z_max=x_espesor, color='gray', alpha=0.3)

    # B) Dibujar el haz incidente - Cilindro Azul (más largo para ver cómo entra/sale)
    draw_cylinder_3d(ax, radius=d_haz/2, z_min=-3, z_max=x_espesor+3, color='blue', alpha=0.15)

    # C) Dibujar el volumen efectivo de interacción - Cilindro Rojo intenso
    draw_cylinder_3d(ax, radius=d_efectivo/2, z_min=0.01, z_max=x_espesor-0.01, color='red', alpha=0.6)

    # Ajustes de los ejes para mantener la proporción geométrica
    limite_xy = max(d_haz, d_blanco) / 2 + 1
    ax.set_xlim(-limite_xy, limite_xy)
    ax.set_ylim(-limite_xy, limite_xy)
    ax.set_zlim(-3, x_espesor + 3)

    # Etiquetas de los ejes
    ax.set_xlabel('Eje X (cm)')
    ax.set_ylabel('Eje Y (cm)')
    ax.set_zlabel('Profundidad Z (cm)')
    ax.set_title('Visualización 3D: Interacción Haz de Fotones vs Blanco', pad=20)

    # Ángulo de cámara inicial (Elevación y Azimut)
    ax.view_init(elev=20, azim=40)

    plt.show()

# Lanzar los controles interactivos
widgets.interact(calcular_interacciones_3d,
                 d_haz=widgets.FloatSlider(value=2.0, min=1.0, max=15.0, step=0.5, description='d_haz (cm):'),
                 d_blanco=widgets.FloatSlider(value=4.0, min=1.0, max=15.0, step=0.5, description='d_blanco (cm):'),
                 x_espesor=widgets.FloatSlider(value=10.0, min=0.5, max=20.0, step=0.5, description='Espesor x (cm):'))
```

### **Problema 2**

Para fotones de 150 keV incidiendo sobre Cu (Z=29, Pa=63.54), la sección eficaz de scattering coherente es de 2.20 $\times$ 10$^{-24}$ cm$^{2}$ y la sección eficaz fotoeléctrica es 8.89 $\times$ 10$^{-24}$ cm$^{2}$. Se sabe además que para esa energía las secciones eficaces Compton (colisión fotón - electrón) de scattering y de transferencia valen 0.363 $\times$ 10$^{-24}$ cm$^{2}$ y 0.081 $\times$ 10$^{-24}$ cm$^{2}$ respectivamente.

Suponiendo un fotón individual que incide sobre una lámina de cobre ($\rho$ = 8.96 g cm$^{-3}$) de 0.5 cm de espesor, hallar la probabilidad de que:

a) el fotón sea dispersado elásticamente.

b) el fotón sea dispersado inelásticamente cediendo parte de su energía a un electrón del medio.

c) el fotón sea totalmente absorbido.

d) Hallar la energía media transferida por colisión.


---
#### Datos iniciales y preparación de las secciones eficaces
Tenemos un fotón de $E = 150 \text{ keV}$ incidiendo en una lámina de Cobre (Cu) con $Z=29$, $P_a = 63.54 \text{ g}$ y $\rho = 8.96 \text{ g/cm}^3$, con un espesor de $x = 0.5 \text{ cm}$.

Primero, calculamos las secciones eficaces macroscópicas y totales. El enunciado nos da la sección eficaz del efecto Compton por *electrón* ($\sigma_{sc}$ y $\sigma_{tr}$). Dado que el scattering incoherente ocurre con todos los electrones del átomo, multiplicamos por $Z$:
*   **Sección eficaz incoherente (Compton por átomo):**
    $\sigma_{inc} = Z \times (\sigma_{sc} + \sigma_{tr}) = 29 \times (0.363 + 0.081) \times 10^{-24} \text{ cm}^2 = \mathbf{12.9 \times 10^{-24} \text{ cm}^2}$.
*   **Sección eficaz fotoeléctrica:** $\sigma_{ph} = \mathbf{8.89 \times 10^{-24} \text{ cm}^2}$.
*   **Sección eficaz de scattering coherente:** $\sigma_{coh} = \mathbf{2.20 \times 10^{-24} \text{ cm}^2}$.

**Sección eficaz total ($\sigma$):**
Sumamos las contribuciones de todos los procesos:
$\sigma = \sigma_{coh} + \sigma_{ph} + \sigma_{inc} = (2.20 + 8.89 + 12.9) \times 10^{-24} \text{ cm}^2 = \mathbf{24.0 \times 10^{-24} \text{ cm}^2}$.

#### Cálculo de la Densidad Atómica ($N$) y el exponente de atenuación**
El número de átomos por cm$^3$ en el Cobre se obtiene mediante el número de Avogadro:
$N = \frac{N_A \cdot \rho}{P_a} = \frac{(6.02 \times 10^{23}) \cdot 8.96}{63.54} = \mathbf{8.49 \times 10^{22} \text{ cm}^{-3}}$.

El exponente de la ley de atenuación para el espesor de 0.5 cm será:
$N \cdot \sigma \cdot x = (8.49 \times 10^{22}) \times (24.0 \times 10^{-24}) \times 0.5 = \mathbf{1.0188}$

---

#### Probabilidades de interacción (Apartados a, b y c)
> **⚠️ Nota metodológica:** Un error común es intentar calcular la probabilidad individual de un proceso usando $P_i = 1 - e^{-N \sigma_i x}$ (que daría resultados incorrectos del 9%, 42% y 31% respectivamente). La forma correcta, rigurosa para blanco grueso, es calcular la *probabilidad total* de que haya una interacción y, a continuación, multiplicarla por el *peso relativo* ($\sigma_i / \sigma$) del proceso específico.

*   **Probabilidad Total de interacción:**
    $P_{tot} = 1 - \exp(-N \sigma x) = 1 - e^{-1.0188} = \mathbf{0.639}$ (es decir, el 63.9% de los fotones interaccionan).

A continuación, calculamos la probabilidad de cada proceso específico $P_i = P_{tot} \times \left(\frac{\sigma_i}{\sigma}\right)$:

**a) Dispersión elástica (Scattering Coherente):**

La proporción del proceso es $\frac{2.20}{24.0} \approx 0.09$.
$P_{coh} = 0.639 \times 0.09 = \mathbf{0.06 \text{ (6\%)}}$.

**b) Dispersión inelástica cediendo energía (Scattering Incoherente / Compton):**

La proporción del proceso es $\frac{12.9}{24.0} \approx 0.54$.
$P_{inc} = 0.639 \times 0.54 = \mathbf{0.34 \text{ (34\%)}}$.

**c) Absorción total (Efecto Fotoeléctrico):**

La proporción del proceso es $\frac{8.89}{24.0} \approx 0.37$.
$P_{ph} = 0.639 \times 0.37 = \mathbf{0.24 \text{ (24\%)}}$.

---

**d) Energía media transferida por colisión ($\bar{E}_{tr}$)**

Para calcular la energía media que el fotón transfiere al medio **cuando colisiona**, debemos ponderar la energía que se transfiere en cada tipo de interacción por la probabilidad relativa ($\sigma_i/\sigma$) de que ocurra dicha interacción:

1.  **Scattering Coherente:** El fotón solo cambia de dirección sin ceder energía. $E_{tr}^{coh} = \mathbf{0 \text{ keV}}$.
2.  **Efecto Fotoeléctrico:** El fotón es absorbido por completo. $E_{tr}^{ph} = \mathbf{150 \text{ keV}}$.
3.  **Efecto Compton:** La energía transferida viene dada por la relación entre las secciones de transferencia y scattering:
    $E_{tr}^{inc} = E_0 \times \left( \frac{\sigma_{tr}}{\sigma_{tr} + \sigma_{sc}} \right) = 150 \times \left( \frac{0.081}{0.363 + 0.081} \right) = \mathbf{27.4 \text{ keV}}$.

**Cálculo ponderado:**
$$\bar{E}_{tr} = \sum \left( E_{tr,i} \times \frac{\sigma_i}{\sigma} \right)$$
$$\bar{E}_{tr} = (0 \cdot 0.09) + (150 \cdot 0.37) + (27.4 \cdot 0.54)$$
$$\bar{E}_{tr} = 0 + 55.5 + 13.7 = \mathbf{69.2 \text{ keV}}$$

*(La suma total confirma que, en promedio, un fotón de 150 keV que colisiona transferirá 69.2 keV a los electrones del Cobre).*

```python
# Datos fijos del Problema 2 (Cobre a 150 keV)
N = 8.49e22  # Densidad atómica (cm^-3)
sigma_coh = 2.20e-24  # cm^2
sigma_ph = 8.89e-24   # cm^2
# Calculamos incoherente como Z * (sc + tr)
sigma_inc = 29 * (0.363 + 0.081) * 1e-24  # 12.876e-24 cm^2
sigma_tot = sigma_coh + sigma_ph + sigma_inc  # ~ 23.97e-24 cm^2

def analizar_probabilidades_cu(x_espesor):
    # 1. Fórmula Correcta: Probabilidad total ponderada
    P_tot = 1 - np.exp(-N * sigma_tot * x_espesor)

    P_corr = {
        'Scatt. Coherente': P_tot * (sigma_coh / sigma_tot),
        'Fotoeléctrico': P_tot * (sigma_ph / sigma_tot),
        'Scatt. Incoherente': P_tot * (sigma_inc / sigma_tot)
    }

    # 2. Fórmula Incorrecta (aplicando exponencial a cada proceso por separado)
    P_incorr = {
        'Scatt. Coherente': 1 - np.exp(-N * sigma_coh * x_espesor),
        'Fotoeléctrico': 1 - np.exp(-N * sigma_ph * x_espesor),
        'Scatt. Incoherente': 1 - np.exp(-N * sigma_inc * x_espesor)
    }

    # --- Interfaz de Texto ---
    print(f"--- Análisis para lámina de Cobre de {x_espesor:.2f} cm ---")
    print(f"Probabilidad TOTAL de que haya interacción: {P_tot*100:.1f}%\n")
    print(f"{'Proceso':<20} | {'Fórmula CORRECTA':<18} | {'Fórmula INCORRECTA':<18}")
    print("-" * 62)
    for proc in P_corr:
        print(f"{proc:<20} | {P_corr[proc]*100:>15.1f} % | {P_incorr[proc]*100:>16.1f} %")

    # --- Gráfico de Barras Comparativo ---
    fig, ax = plt.subplots(figsize=(9, 5))

    etiquetas = list(P_corr.keys())
    valores_corr = [P_corr[k]*100 for k in etiquetas]
    valores_incorr = [P_incorr[k]*100 for k in etiquetas]

    x = np.arange(len(etiquetas))
    ancho = 0.35

    # Dibujar barras
    ax.bar(x - ancho/2, valores_corr, ancho, label=r'Correcta: $P_{tot} \cdot (\sigma_i / \sigma)$', color='#2ca02c')
    ax.bar(x + ancho/2, valores_incorr, ancho, label=r'Incorrecta: $1 - e^{-N \sigma_i x}$', color='#d62728', alpha=0.7)

    # Configuración estética
    ax.set_ylabel('Probabilidad (%)')
    ax.set_title(f'Impacto del cálculo en blanco grueso (Espesor = {x_espesor} cm)')
    ax.set_xticks(x)
    ax.set_xticklabels(etiquetas)
    ax.legend(loc='upper left')
    ax.grid(axis='y', linestyle=':', alpha=0.7)

    # Ajuste dinámico del eje Y
    ax.set_ylim(0, max(max(valores_corr), max(valores_incorr)) + 15)

    plt.tight_layout()
    plt.show()

# Lanzar los controles interactivos (valor por defecto 0.5 cm como en el problema)
widgets.interact(analizar_probabilidades_cu,
                 x_espesor=widgets.FloatSlider(value=0.5, min=0.01, max=4.0, step=0.05, description='Espesor x (cm):', continuous_update=False))
```

### **Problema 3**

La sección eficaz total de scattering Compton por electrón libre viene dada por la expresión de Klein-Nishina:

$$ \sigma_{KN} = 2\pi r_0^2 \left[ \frac{1 + \gamma}{\gamma^2} \left( \frac{2(1 + \gamma)}{1 + 2\gamma} - \frac{\ln(1 + 2\gamma)}{\gamma} \right) + \frac{\ln(1 + 2\gamma)}{2\gamma} - \frac{1 + 3\gamma}{(1 + 2\gamma)^2} \right] $$

donde $\gamma = \frac{E}{m_e c^2}$.

**Estudiar la validez del efecto Compton** para describir el scattering incoherente $\sigma_{incoh}$ de fotones en los átomos de Carbono (C) y Plomo (Pb).

Para ello, **representar gráficamente** la diferencia:

$$ \frac{Z \cdot \sigma_{KN} - \sigma_{incoh}}{\langle\sigma\rangle} $$

frente a la energía del fotón, donde

$$ \langle\sigma\rangle = \frac{1}{2} ( Z \cdot \sigma_{KN} + \sigma_{incoh}) $$

¿Qué **proceso de interacción domina** para las energías en las que las secciones eficaces Compton e incoherente son sensiblemente diferentes?

*Nota: Las secciones eficaces atómicas pueden encontrarse en la base de datos de la web del NIST: [http://physics.nist.gov/PhysRefData/Xcom/html/xcom1.html](http://physics.nist.gov/PhysRefData/Xcom/html/xcom1.html).*

Para comprender el alcance y la validez del modelo de Klein-Nishina en el cálculo del scattering incoherente, es necesario analizar los fenómenos físicos subyacentes que diferencian a un electrón libre de uno ligado a un átomo.

#### **1. El Efecto de Ligadura Atómica y la Supresión del Scattering**
La sección eficaz de Klein-Nishina ($\sigma_{KN}$) se deriva bajo la hipótesis de que el fotón interacciona con un **electrón libre y en reposo**. En esta aproximación, la sección eficaz atómica se estima simplemente sumando la contribución de cada uno de los $Z$ electrones de manera independiente:
$$ \sigma_{incoh} \approx Z \cdot \sigma_{KN} $$

Sin embargo, en un átomo real, los electrones están **ligados** en diferentes capas con energías de ligadura $E_b$ específicas.
* **A bajas energías ($E \lesssim E_b$):** El fotón no puede transferir cualquier cantidad de energía al electrón debido a las restricciones de los niveles energéticos cuantizados del átomo. El proceso de scattering incoherente se ve fuertemente **suprimido** (lo que experimentalmente se modela multiplicando por la función de dispersión incoherente $S(q, Z)$, que tiende a cero a bajas energías). Por tanto, la sección eficaz real $\sigma_{inc}$ es mucho menor que la predicción teórica: $\sigma_{inc} \ll Z\sigma_{KN}$.
* **A altas energías ($E \gg E_b$):** La energía del fotón es tan superior a la energía de ligadura de los electrones (incluso de la capa K) que estos pueden considerarse efectivamente "libres y en reposo". En este régimen, la aproximación teórica converge perfectamente con la realidad: $\sigma_{inc} \approx Z\sigma_{KN}$.


#### **2. Análisis por Elemento**

* **Carbono ($Z = 6$, elemento ligero):**
  Al tener electrones débilmente ligados (la energía de ligadura de la capa K es de apenas $0.28\text{ keV}$), la aproximación de electrón libre se vuelve excelente ($<5\%$ de diferencia) a partir de energías muy bajas, aproximadamente **$E > 20\text{ keV}$**.
  Por debajo de esta energía, la diferencia se dispara rápidamente hacia el $200\%$. Sin embargo, en esta región de bajas energías el proceso dominante es, con gran diferencia, el **efecto fotoeléctrico** ($\sigma_{ph} \gg \sigma_{inc}$).

* **Plomo ($Z = 82$, elemento pesado):**
  Al tener electrones muy fuertemente ligados (la energía de ligadura de la capa K es de $88\text{ keV}$), los efectos de ligadura se hacen notar hasta energías mucho más elevadas. La aproximación no empieza a ser buena ($<5\%$ de diferencia) hasta que superamos aproximadamente los **$E > 500\text{ keV}$** (0.5 MeV).
  Al igual que en el Carbono, en la zona donde la aproximación Compton falla (bajas energías), el **efecto fotoeléctrico** es el proceso de interacción absolutamente dominante en el Plomo.

#### **3. Conclusión sobre la Validez de la Aproximación**
A efectos prácticos de física de radiaciones y blindajes, **la aproximación de describir el scattering incoherente mediante $Z \cdot \sigma_{KN}$ es plenamente válida**.

Esto se debe a una coincidencia:
1. En el rango de energías donde la aproximación es **incorrecta** (bajas energías), el error cometido es irrelevante porque el **efecto fotoeléctrico absorbe prácticamente el 100% de los fotones**, haciendo que la contribución del scattering a la atenuación total sea despreciable.
2. En el rango de energías donde el **scattering incoherente es el proceso dominante**, la energía del fotón es tan alta que los electrones se comportan como libres, haciendo que la aproximación teórica de Klein-Nishina sea **extremadamente precisa**.


```python
# ---------------------------------------------------------
# CONSTANTES FÍSICAS
# ---------------------------------------------------------
r0 = 2.8179e-15  # Radio clásico del electrón (m)
barn = 1e-28     # Conversión de m^2 a barns
mec2 = 0.511     # Energía en reposo del electrón (MeV)
NA = 6.0221415e23 # Número de Avogadro

def sigma_KN(E_MeV):
    """
    Fórmula teórica de Klein-Nishina para la sección eficaz total
    de scattering Compton por electrón libre (devuelve barns).
    """
    E_MeV = np.where(E_MeV == 0, 1e-10, E_MeV)
    gamma = E_MeV / mec2

    term1 = ((1 + gamma) / gamma**2) * (2 * (1 + gamma) / (1 + 2*gamma) - np.log(1 + 2*gamma) / gamma)
    term2 = np.log(1 + 2*gamma) / (2 * gamma)
    term3 = (1 + 3*gamma) / (1 + 2*gamma)**2

    sigma_m2 = 2 * np.pi * r0**2 * (term1 + term2 - term3)
    return sigma_m2 / barn

def leer_y_convertir_datos_nist(filepath, A):
    """
    Lee los datos del NIST en cm^2/g y los convierte a barns/átomo
    usando la masa atómica (A) del elemento.
    """
    energies, incoherents, photoelectrics = [], [], []

    # Factor de conversión de cm^2/g a barns/átomo
    # (A / NA) * 10^24 = A * 1.66054
    factor_conversion = A * 1.66054125

    try:
        with open(filepath, 'r') as f:
            for line in f:
                parts = line.strip().split()
                if len(parts) == 8:
                    try:
                        e = float(parts[0])
                        # Extraemos y aplicamos la conversión de unidades
                        inc = float(parts[2]) * factor_conversion
                        ph = float(parts[3]) * factor_conversion

                        energies.append(e)
                        incoherents.append(inc)
                        photoelectrics.append(ph)
                    except ValueError:
                        continue

        return np.array(energies), np.array(incoherents), np.array(photoelectrics)
    except FileNotFoundError:
        print(f"Error: No se encontró el archivo '{filepath}'.")
        return None, None, None

# ---------------------------------------------------------
# FUNCIÓN GENERADORA DE GRÁFICOS
# ---------------------------------------------------------
def analizar_validez_compton_ficheros(archivo_txt, simbolo, Z, A, ylim_sigma):
    """
    Analiza la validez de la aproximación Compton convirtiendo las unidades correctamente.
    """
    E, s_inc, s_ph = leer_y_convertir_datos_nist(archivo_txt, A)

    if E is None:
        return

    Z_s_KN = Z * sigma_KN(E)

    # ====================================================
    # GRÁFICO 1: Secciones Eficaces (barns/átomo) vs Energía
    # ====================================================
    plt.figure(figsize=(8, 6))
    plt.plot(E, Z_s_KN, label=r'Z * $\sigma_{KN}$ (teórica)', color='#4c72b0', linewidth=2.5)
    plt.plot(E, s_inc, label='Incoherente (NIST real)', color='#c44e52', linewidth=2.5, linestyle='--')
    plt.plot(E, s_ph, label='Fotoeléctrico (NIST real)', color='#8ebad9', linewidth=2.5)

    plt.xscale('log')
    plt.yscale('log')
    plt.xlim(1e-3, 1e6)
    plt.ylim(1e-4, ylim_sigma)

    plt.xlabel('Energía (MeV)', fontsize=12)
    plt.ylabel(r'$\sigma$ (barn/átomo)', fontsize=12)
    plt.title(f'{simbolo} - Secciones eficaces', fontsize=14, fontweight='bold')
    plt.grid(True, which="both", ls="-", alpha=0.3)
    plt.legend(loc='upper right', fontsize=11)

    plt.text(0.1, 0.85, simbolo, transform=plt.gca().transAxes,
             fontsize=30, fontweight='bold', bbox=dict(facecolor='white', alpha=0.8, edgecolor='gray'))

    plt.tight_layout()
    plt.show()

    # ====================================================
    # GRÁFICO 2: Diferencia Relativa (%)
    # ====================================================
    diferencia_pct = (Z_s_KN - s_inc) / (0.5 * (Z_s_KN + s_inc)) * 100

    plt.figure(figsize=(6, 6))
    plt.plot(E, diferencia_pct, 'D', color='darkblue', markersize=5,
             label=r'$(Z \cdot \sigma_{KN} - \sigma_{inc}) / <\sigma>$')

    plt.xscale('log')
    plt.xlim(1e-3, 1e1)
    plt.ylim(-5, 200)
    plt.gca().yaxis.set_major_formatter(plt.FuncFormatter(lambda y, _: f'{int(y)}%'))

    plt.xlabel('Energía (MeV)', fontsize=12)
    plt.ylabel('Diferencia Compton - Incoherente', fontsize=12)
    plt.grid(True, which="both", ls="-", alpha=0.3)
    plt.legend()

    plt.text(0.7, 0.7, simbolo, transform=plt.gca().transAxes, fontsize=30, fontweight='bold')

    plt.tight_layout()
    plt.show()

# ---------------------------------------------------------
# EJECUCIÓN
# ---------------------------------------------------------
# Para Carbono (Z=6, A=12.011, límite Y=10^3)
analizar_validez_compton_ficheros('data/C_cross-sections-NIST.txt', simbolo='C', Z=6, A=12.011, ylim_sigma=1e4)

# Para Plomo (Z=82, A=207.2, límite Y=10^4)
analizar_validez_compton_ficheros('data/Pb_cross-sections-NIST.txt', simbolo='Pb', Z=82, A=207.2, ylim_sigma=1e4)
```

### Problema 4:

Un haz de rayos $\gamma$ de $150 \text{ keV}$ que contiene un total de $10^{15}$ fotones incide sobre una lámina de cierto material con un espesor másico de $0.85 \text{ g}\cdot\text{cm}^{-2}$ [1, 2]. Se observa experimentalmente que:

* **a)** El $6.53\%$ de los fotones incidentes son absorbidos [1, 2].
* **b)** Un $1.61\%$ de los fotones salen con la misma energía de incidencia, pero habiendo modificado su dirección de movimiento [1, 2].
* **c)** El $9.10\%$ de los fotones salen también con dirección modificada, pero habiendo perdido parte de su energía, registrándose un valor medio para esta pérdida de energía de $123 \text{ keV}$ [1-3].

Los restantes fotones consiguen atravesar la lámina metálica sin sufrir ninguna clase de interacción [1, 3]. Suponiendo una aproximación de colisión única (ningún fotón sufre más de una interacción al atravesar la lámina), calcule:

* **a)** Las **secciones eficaces** (en barns) correspondientes a los procesos de scattering coherente ($\sigma_{coh}$), scattering incoherente ($\sigma_{inc}$) y absorción fotoeléctrica ($\sigma_{ph}$) [3].
* **b)** La **energía total (expresada en Julios)** que el haz de fotones transfiere a los electrones del medio material de la lámina [3].

**DATO:** Densidad atómica del material: $\rho_A = 9.48 \times 10^{21} \text{ átomos/g}$ [3].

---
Este problema se resuelve aplicando la teoría de atenuación de haces de fotones en láminas delgadas bajo la aproximación de colisión única (espesor delgado).

---

#### a) Cálculo de las Secciones Eficaces ($\sigma_{coh}$, $\sigma_{inc}$ y $\sigma_{ph}$)

##### **1. Probabilidad Total de Interacción ($P$)**
Dado que los tres procesos de interacción (absorción fotoeléctrica, scattering coherente y scattering incoherente) son independientes y mutuamente excluyentes para un único fotón, la probabilidad total de que un fotón sufra alguna interacción al atravesar la lámina es la suma de las probabilidades experimentales observadas:

$$ P = P_{ph} + P_{coh} + P_{inc} $$

Sustituyendo los datos del enunciado:
$$ P = 6.53\% + 1.61\% + 9.10\% = 0.0653 + 0.0161 + 0.0910 = 0.1724 \quad (17.24\%) $$

##### **2. Espesor Atómico por Unidad de Superficie ($N_x$)**
Necesitamos calcular el número de átomos que el haz de fotones encuentra por cada $\text{cm}^2$ de sección transversal al atravesar la lámina. Esto se obtiene multiplicando la densidad atómica del material ($\rho_A$) por el espesor másico ($t$ o $\rho x$):

$$ N_x = \rho_A \cdot t $$
$$ N_x = \left(9.48 \times 10^{21} \text{ átomos/g}\right) \times \left(0.85 \text{ g/cm}^2\right) = 8.058 \times 10^{21} \text{ átomos/cm}^2 $$

##### **3. Sección Eficaz Total de Interacción ($\sigma$)**
La atenuación de un haz de fotones sigue una ley exponencial. La probabilidad de que un fotón **no** interaccione (es decir, que atraviese la lámina limpio) es $1 - P = e^{-N_x \cdot \sigma}$. Despejando la sección eficaz total de interacción ($\sigma$) de esta expresión:

$$ \sigma = -\frac{1}{N_x} \ln(1 - P) $$

Sustituyendo los valores obtenidos:
$$ \sigma = -\frac{\ln(1 - 0.1724)}{8.058 \times 10^{21} \text{ cm}^{-2}} \approx 2.3486 \times 10^{-23} \text{ cm}^2 $$

Sabiendo que $1 \text{ barn} = 10^{-24} \text{ cm}^2$:
$$ \sigma \approx 23.5 \text{ barns} $$

##### **4. Secciones Eficaces Individuales ($\sigma_i$)**
Dado que ningún fotón interacciona más de una vez (lámina delgada), la probabilidad de que ocurra un proceso específico $i$ es proporcional a la contribución de su sección eficaz respecto a la total ($\sigma_i = \sigma \cdot \frac{P_i}{P}$) [1]:

*   **Absorción Fotoeléctrica ($\sigma_{ph}$):**
    $$ \sigma_{ph} = \sigma \cdot \frac{P_{ph}}{P} = 2.35 \times 10^{-23} \text{ cm}^2 \times \frac{0.0653}{0.1724} \approx \mathbf{8.89 \times 10^{-24} \text{ cm}^2 \quad (8.89 \text{ barns})} $$

*   **Scattering Coherente ($\sigma_{coh}$):**
    $$ \sigma_{coh} = \sigma \cdot \frac{P_{coh}}{P} = 2.35 \times 10^{-23} \text{ cm}^2 \times \frac{0.0161}{0.1724} \approx \mathbf{2.20 \times 10^{-24} \text{ cm}^2 \quad (2.20 \text{ barns})} $$

*   **Scattering Incoherente ($\sigma_{inc}$):**
    $$ \sigma_{inc} = \sigma \cdot \frac{P_{inc}}{P} = 2.35 \times 10^{-23} \text{ cm}^2 \times \frac{0.0910}{0.1724} \approx \mathbf{1.24 \times 10^{-23} \text{ cm}^2 \quad (12.4 \text{ barns})} $$

---

#### b) Energía Total Transferida a los Electrones ($E_{transf}$)

La energía total transferida por el haz de radiación a la lámina depende exclusivamente de los procesos que ceden energía cinética a los electrones del medio:
1. En el **efecto fotoeléctrico**, el fotón desaparece por completo cediendo toda su energía de incidencia ($E_0 = 150 \text{ keV}$) [1].
2. En el **scattering incoherente (Compton)**, el fotón se dispersa desviado, pero cede una fracción de su energía al electrón de retroceso. El enunciado nos da la pérdida media directamente: $\bar{E}_{perdida} = 123 \text{ keV}$ [1].
3. En el **scattering coherente (Rayleigh)**, la colisión es elástica (el fotón solo cambia de dirección sin perder energía), por lo que la energía transferida al medio es nula [1].

Para un haz de $n = 10^{15}$ fotones incidentes, la energía total transferida (en $\text{keV}$) se calcula ponderando las pérdidas por las probabilidades de que ocurran dichos sucesos [1]:

$$ E_{transf} = n \cdot \left( P_{ph} \cdot E_0 + P_{inc} \cdot \bar{E}_{perdida} \right) $$

Sustituyendo los valores numéricos:
$$ E_{transf} = 10^{15} \cdot \left( 0.0653 \cdot 150 \text{ keV} + 0.0910 \cdot 123 \text{ keV} \right) $$
$$ E_{transf} = 10^{15} \cdot \left( 9.795 \text{ keV} + 11.193 \text{ keV} \right) = 2.0988 \times 10^{16} \text{ keV} $$

##### **Conversión a Julios ($J$):**
Utilizando la relación de conversión elemental ($1 \text{ eV} = 1.6022 \times 10^{-19} \text{ J}$, por lo que $1 \text{ keV} = 1.6022 \times 10^{-16} \text{ J}$) [1]:

$$ E_{transf} = \left(2.0988 \times 10^{16} \text{ keV}\right) \times \left(1.6022 \times 10^{-16} \text{ J/keV}\right) \approx \mathbf{3.36 \text{ Julios}} $$


```python
# 1. DATOS DE ENTRADA
n_fotones = 1e15             # Número total de fotones incidentes
E0_keV = 150.0               # Energía inicial de los fotones (keV)
t_espesor_masico = 0.85      # Espesor másico t (g/cm^2)
rho_A = 9.48e21              # Densidad atómica (átomos/g)

# Probabilidades experimentales de interacción (%)
P_ph_val = 0.0653            # Absorción fotoeléctrica (6.53%)
P_coh_val = 0.0161           # Scattering coherente (1.61%)
P_inc_val = 0.0910           # Scattering incoherente (9.10%)
E_perdida_inc_keV = 123.0    # Pérdida media de energía Compton (keV)

# Conversión de unidades
barn = 1e-24                 # 1 barn = 10^-24 cm^2
eV_to_J = 1.602176634e-19    # Conversión de eV a Julios
keV_to_J = eV_to_J * 1000     # Conversión de keV a Julios

# =====================================================================
# DESARROLLO MATEMÁTICO
# =====================================================================

# Espesor atómico por unidad de superficie N*x (átomos/cm^2)
Nx = rho_A * t_espesor_masico

# Probabilidad de interacción total acumulada
P_tot = P_ph_val + P_coh_val + P_inc_val

def calcular_problema_4(P_total):
    # Sección eficaz total (cm^2) -> sigma = -ln(1 - P) / (N*x)
    sigma_cm2 = -np.log(1 - P_total) / Nx
    sigma_barns = sigma_cm2 / barn

    # Secciones eficaces individuales: sigma_i = sigma * (P_i / P_total)
    sigma_ph = sigma_barns * (P_ph_val / P_total)
    sigma_coh = sigma_barns * (P_coh_val / P_total)
    sigma_inc = sigma_barns * (P_inc_val / P_total)

    # Energía total transferida en Julios
    # E_transf = n * (P_ph * E0 + P_inc * E_tr_inc)
    # El scattering coherente no transfiere energía (dispersión elástica)
    E_transf_keV = n_fotones * (P_ph_val * E0_keV + P_inc_val * E_perdida_inc_keV)
    E_transf_J = E_transf_keV * keV_to_J

    # Imprimir resultados
    print(f"Probabilidad total (P): {P_total*100:.2f}%")
    print(f"Espesor atómico (N.x):  {Nx:.4e} átomos/cm²")
    print(f"Sección eficaz total:   {sigma_barns:.2f} barns")
    print(f"  └─ Fotoeléctrica (σ_ph):  {sigma_ph:.2f} barns")
    print(f"  └─ Coherente (σ_coh):     {sigma_coh:.2f} barns")
    print(f"  └─ Incoherente (σ_inc):   {sigma_inc:.2f} barns")
    print(f"Energía total transferida:  {E_transf_J:.2f} Julios")
    print("-" * 40)

# Ejecutar cálculos
calcular_problema_4(P_tot)
```

### Problema 5:

Hallar la energía media transferida por colisión $\bar{E}_{tr}$ para rayos $\gamma$ de $0.1$, $5$ y $100\text{ MeV}$ sobre plomo. El cociente $\sigma_{tr}/\sigma_{sc}$ en el scattering Compton vale $0.160$, $1.69$ y $3.82$ para $0.1$, $5$ y $100\text{ MeV}$ respectivamente.

Además, se disponen de los siguientes datos de sección eficaz de interacción (en unidades de $10^{-24}\text{ cm}^2$ o barns):

| Proceso | $0.1\text{ MeV}$ | $5.0\text{ MeV}$ | $100\text{ MeV}$ |
| :--- | :---: | :---: | :---: |
| **Scattering Coherente ($\sigma_{coh}$)** | $67.6$ | $0.037$ | — |
| **Efecto Fotoeléctrico ($\sigma_{ph}$)** | $1777$ | $0.410$ | $0.013$ |
| **Scattering Incoherente ($\sigma_{inc}$)** | $34.4$ | $6.81$ | $0.679$ |
| **Creación de Pares ($\sigma_{par}$)** | — | $7.29$ | $31.5$ |


Para hallar la energía media transferida por colisión ($\bar{E}_{tr}$) en un medio monoatómico, debemos ponderar la energía media transferida en cada proceso individual por la probabilidad de ocurrencia de dicho proceso (que es proporcional a su sección eficaz respecto a la sección eficaz total de interacción, $\sigma_T$) [5]:

$$ \bar{E}_{tr} = \sum_{i} \bar{E}_{tr}^i \frac{\sigma_i}{\sigma_T} $$

Sustituyendo los procesos físicos involucrados, la ecuación general queda expresada como:

$$ \bar{E}_{tr} = \bar{E}_{tr}^{coh} \frac{\sigma_{coh}}{\sigma_T} + \bar{E}_{tr}^{ph} \frac{\sigma_{ph}}{\sigma_T} + \bar{E}_{tr}^{inc} \frac{\sigma_{inc}}{\sigma_T} + \bar{E}_{tr}^{par} \frac{\sigma_{par}}{\sigma_T} $$

---

#### **Análisis de la Energía Transferida por Proceso**

1. **Scattering Coherente (Rayleigh):**
   Es una dispersión elástica pura del fotón con el átomo completo. No se produce ionización ni excitación electrónica, por lo que la energía transferida al medio es nula:
   $$ \bar{E}_{tr}^{coh} = 0 $$

2. **Efecto Fotoeléctrico:**
   El fotón es absorbido completamente. En una aproximación simplificada, se considera que toda la energía del fotón se transfiere localmente a los electrones (despreciando pérdidas por radiación de fluorescencia):
   $$ \bar{E}_{tr}^{ph} \approx E_0 $$

3. **Scattering Incoherente (Compton):**
   El fotón es desviado inelásticamente por un electrón. La sección eficaz total de scattering Compton ($\sigma_{KN}$) es la suma de la sección eficaz de dispersión ($\sigma_{sc}$) y la de transferencia de energía ($\sigma_{tr}$):
   $$ \sigma_{KN} = \sigma_{tr} + \sigma_{sc} $$
   
   La fracción de la energía inicial $E_0$ que se transfiere de media al electrón de retroceso es exactamente el cociente $\sigma_{tr}/\sigma_{KN}$. Podemos reescribir esta fracción en términos del cociente $\sigma_{tr}/\sigma_{sc}$ dado en el enunciado:
   
   $$ \bar{E}_{tr}^{inc} = E_0 \cdot \frac{\sigma_{tr}}{\sigma_{KN}} = E_0 \cdot \frac{\sigma_{tr}}{\sigma_{tr} + \sigma_{sc}} = E_0 \cdot \left(1 + \frac{\sigma_{sc}}{\sigma_{tr}}\right)^{-1} $$

4. **Creación de Pares:**
   El fotón desaparece cerca del núcleo atómico dando lugar a un par electrón-positrón. La energía umbral para que ocurra el proceso es de $2 m_e c^2 \approx 1.022\text{ MeV}$ (la energía necesaria para crear las masas en reposo de ambas partículas). Toda la energía excedente se transfiere al medio en forma de energía cinética de la pareja creada:
   $$ \bar{E}_{tr}^{par} = E_0 - 2m_e c^2 = E_0 - 1.022\text{ MeV} $$

---

#### **Cálculos Numéricos por Energía**

##### **Caso 1: $E_0 = 0.1\text{ MeV}$ (Baja energía)**
*   **Sección eficaz total:** $\sigma_T = 67.6 \text{ (coh)} + 1777 \text{ (ph)} + 34.4 \text{ (inc)} = 1879 \times 10^{-24}\text{ cm}^2$.
*   **Compton:** Con $\frac{\sigma_{tr}}{\sigma_{sc}} = 0.160$, la fracción Compton transferida es $(1 + 1/0.160)^{-1} = 0.1379$. La energía Compton transferida es $\bar{E}_{tr}^{inc} = 0.1\text{ MeV} \times 0.1379 = 0.0138\text{ MeV}$.
*   **Contribuciones individuales ($\bar{E}_{tr}^i \frac{\sigma_i}{\sigma_T}$):**
    *   Fotoeléctrico: $0.1\text{ MeV} \times \frac{1777}{1879} \approx 0.0946\text{ MeV}$.
    *   Incoherente: $0.0138\text{ MeV} \times \frac{34.4}{1879} \approx 0.00025\text{ MeV}$.
*   **Energía media transferida total:**
    $$ \bar{E}_{tr} = 0.0946 + 0.00025 \approx \mathbf{0.0949\text{ MeV}} \quad (\text{o } 94.9\text{ keV}) $$

> 💡 **Nota sobre el Efecto de la Fluorescencia:**
> Si se consulta, por ejemplo, la Tabla A-4i en [Johns and Cunningham](https://ucm.on.worldcat.org/search/detail/1454697710?queryString=The%20Physics%20of%20Radiology&idDetect=true&citeDetect=true&clusterResults=false&groupVariantRecords=false&newsArticles=off&bookReviews=off$0), el valor reportado para esta energía es de **$0.0362\text{ MeV}$**. Esta diferencia se debe a que la aproximación simple $\bar{E}_{tr}^{ph} \approx E_0$ ignora que en elementos pesados como el Plomo ($E_K \approx 88\text{ keV}$), una parte sustancial de la energía del efecto fotoeléctrico se emite como rayos X de fluorescencia característicos que escapan del medio local sin transferir su energía a los electrones. Al restar estas pérdidas de radiación secundaria, la energía real transferida disminuye notablemente.

---

##### **Caso 2: $E_0 = 5.0\text{ MeV}$ (Energía media)**
*   **Sección eficaz total:** $\sigma_T = 0.037 + 0.410 + 6.81 + 7.29 = 14.55 \times 10^{-24}\text{ cm}^2$ [4].
*   **Compton:** Con $\frac{\sigma_{tr}}{\sigma_{sc}} = 1.69$, la fracción Compton transferida es $(1 + 1/1.69)^{-1} = 0.6283$. La energía Compton transferida es $\bar{E}_{tr}^{inc} = 5.0\text{ MeV} \times 0.6283 = 3.141\text{ MeV}$.
*   **Creación de pares:** $\bar{E}_{tr}^{par} = 5.0 - 1.022 = 3.978\text{ MeV}$.
*   **Contribuciones individuales ($\bar{E}_{tr}^i \frac{\sigma_i}{\sigma_T}$):**
    *   Fotoeléctrico: $5.0\text{ MeV} \times \frac{0.410}{14.55} \approx 0.141\text{ MeV}$.
    *   Incoherente: $3.141\text{ MeV} \times \frac{6.81}{14.55} \approx 1.47\text{ MeV}$.
    *   Creación de pares: $3.978\text{ MeV} \times \frac{7.29}{14.55} \approx 1.99\text{ MeV}$.
*   **Energía media transferida total:**
    $$ \bar{E}_{tr} = 0.141 + 1.47 + 1.99 \approx \mathbf{3.60\text{ MeV}} $$

---

##### **Caso 3: $E_0 = 100\text{ MeV}$ (Alta energía)**
*   **Sección eficaz total:** $\sigma_T = 0.013 + 0.679 + 31.5 = 32.19 \times 10^{-24}\text{ cm}^2$.
*   **Compton:** Con $\frac{\sigma_{tr}}{\sigma_{sc}} = 3.82$, la fracción Compton transferida es $(1 + 1/3.82)^{-1} = 0.7925$. La energía Compton transferida es $\bar{E}_{tr}^{inc} = 100\text{ MeV} \times 0.7925 = 79.25\text{ MeV}$.
*   **Creación de pares:** $\bar{E}_{tr}^{par} = 100 - 1.022 = 98.98\text{ MeV}$.
*   **Contribuciones individuales ($\bar{E}_{tr}^i \frac{\sigma_i}{\sigma_T}$):**
    *   Fotoeléctrico: $100\text{ MeV} \times \frac{0.013}{32.19} \approx 0.0404\text{ MeV}$.
    *   Incoherente: $79.25\text{ MeV} \times \frac{0.679}{32.19} \approx 1.67\text{ MeV}$.
    *   Creación de pares: $98.98\text{ MeV} \times \frac{31.5}{32.19} \approx 96.9\text{ MeV}$.
*   **Energía media transferida total:**
    $$ \bar{E}_{tr} = 0.0404 + 1.67 + 96.9 \approx \mathbf{98.6\text{ MeV}} $$

```python
# =====================================================================
# CONFIGURACIÓN DE DATOS DEL PROBLEMA
# =====================================================================
energias = np.array([0.1, 5.0, 100.0])  # MeV

# Secciones eficaces en barns (10^-24 cm^2)
s_coh = np.array([67.6, 0.037, 0.0])
s_ph  = np.array([1777.0, 0.410, 0.013])
s_inc = np.array([34.4, 6.81, 0.679])
s_par = np.array([0.0, 7.29, 31.5])

# Relaciones Compton dadas por el enunciado (sigma_tr / sigma_sc)
r_tr_sc = np.array([0.160, 1.69, 3.82])

# =====================================================================
# CÁLCULO DE LAS ENERGÍAS MEDIAS TRANSFERIDAS POR PROCESO
# =====================================================================
# 1. Sección eficaz total (barns)
s_total = s_coh + s_ph + s_inc + s_par

# 2. Energía transferida en efecto fotoeléctrico (Aprox. simple: Eph = E0)
E_tr_ph = energias

# 3. Energía transferida en efecto Compton (incoherente)
# Fracción = sigma_tr / (sigma_tr + sigma_sc) = (1 + sigma_sc/sigma_tr)^-1
fraccion_compton = 1.0 / (1.0 + (1.0 / r_tr_sc))
E_tr_inc = energias * fraccion_compton

# 4. Energía transferida en creación de pares (E0 - 2*me*c^2)
E_tr_par = np.maximum(0.0, energias - 1.022)

# =====================================================================
# CONTRIBUCIONES PARTICULARES A LA ENERGÍA TOTAL
# =====================================================================
contrib_ph  = (s_ph / s_total) * E_tr_ph
contrib_inc = (s_inc / s_total) * E_tr_inc
contrib_par = (s_par / s_total) * E_tr_par

# Energía total transferida (MeV)
E_tr_total = contrib_ph + contrib_inc + contrib_par

# =====================================================================
# GENERACIÓN DE TABLA DE RESULTADOS (FORMATO PANDAS)
# =====================================================================
resultados = pd.DataFrame({
    'Energía del fotón (MeV)': energias,
    'S.E. Total (barn)': s_total,
    'Contrib. Fotoeléctrico (MeV)': contrib_ph,
    'Contrib. Incoherente (MeV)': contrib_inc,
    'Contrib. Creación Pares (MeV)': contrib_par,
    'E_tr Total Calculada (MeV)': E_tr_total,
})

# Mostrar la tabla formateada con 4 decimales
print("=== RESOLUCIÓN NUMÉRICA ===")
display(resultados.round(4))
```

### Problema 6

Un haz de rayos X de $50\text{ keV}$ incide sobre Calcio ($Z=20$, $A=40$). La sección eficaz de scattering coherente vale $\sigma_{coh} = 6.0 \times 10^{-24}\text{ cm}^2$ y la de efecto fotoeléctrico es $\sigma_{ph} = 50.8 \times 10^{-24}\text{ cm}^2$.

Suponiendo que el scattering incoherente pueda ser descrito de forma aproximada por el efecto Compton para el que, a $50\text{ keV}$, las secciones eficaces por electrón de scattering y transferencia valen $\sigma_{sc} = 0.52 \times 10^{-24}\text{ cm}^2$ y $\sigma_{tr} = 0.04 \times 10^{-24}\text{ cm}^2$, calcule:

* **a)** La **energía media transferida** por colisión ($\bar{E}_{tr}$).
* **b)** La **energía media absorbida** por colisión ($\bar{E}_{ab}$).

Para resolver este ejercicio debemos calcular cómo se distribuye la energía en promedio cada vez que un fotón experimenta una colisión en el medio (Calcio).

#### **1. Definición de la Sección Eficaz Total ($\sigma_T$)**
La sección eficaz total de interacción por átomo es la suma de las secciones eficaces de los tres procesos posibles a esta energía: el efecto fotoeléctrico, el scattering coherente y el scattering incoherente (Compton).

Como las secciones eficaces del efecto Compton se proporcionan **por electrón libre**, la sección eficaz atómica incoherente aproximada se obtiene multiplicando la contribución de un único electrón por el número atómico del Calcio ($Z = 20$):

$$ \sigma_{inc} \approx Z \cdot \sigma_{KN} = Z \cdot (\sigma_{sc} + \sigma_{tr}) $$

Sustituyendo los valores numéricos:
$$ \sigma_{inc} \approx 20 \times (0.52 + 0.04) \times 10^{-24}\text{ cm}^2 = 11.2 \times 10^{-24}\text{ cm}^2 \quad (11.2\text{ barns}) $$

La sección eficaz total atómica ($\sigma_T$) es:
$$ \sigma_T = \sigma_{ph} + \sigma_{coh} + \sigma_{inc} $$
$$ \sigma_T = (50.8 + 6.0 + 11.2) \times 10^{-24}\text{ cm}^2 = 68.0 \times 10^{-24}\text{ cm}^2 \quad (68.0\text{ barns}) $$

---

#### **2. Cálculo de la Energía Media Transferida por Colisión ($\bar{E}_{tr}$)**
La energía media transferida al medio por colisión se define como el promedio ponderado de la energía transferida en cada proceso individual por la probabilidad de ocurrencia de dicho proceso (que equivale a $\sigma_i / \sigma_T$):

$$ \bar{E}_{tr} = \sum_{i} \bar{E}_{tr}^i \frac{\sigma_i}{\sigma_T} $$

Analizando la transferencia de energía de cada proceso:
*   **Scattering Coherente (Rayleigh):** Al ser una dispersión elástica sin ionización, no se transfiere energía a los electrones del medio: $\bar{E}_{tr}^{coh} = 0$.
*   **Efecto Fotoeléctrico:** El fotón se absorbe por completo, por lo que (despreciando fluorescencia secundaria) transfiere toda su energía cinética de forma local: $\bar{E}_{tr}^{ph} \approx E_0 = 50\text{ keV}$.
*   **Scattering Incoherente (Compton):** En cada interacción con un electrón, la energía media transferida al electrón de retroceso viene dada por la fracción de transferencia del proceso: $\bar{E}_{tr}^{inc} = E_0 \frac{\sigma_{tr}}{\sigma_{KN}}$.
*   **Creación de pares:** La energía de los fotones incidentes es menor que el umbral necesario para que este proceso tenga lugar ($2m_e$).

Ponderando estos términos:

$$ \bar{E}_{tr} = E_0 \frac{\sigma_{ph}}{\sigma_T} + \left(E_0 \frac{\sigma_{tr}}{\sigma_{KN}}\right) \frac{Z \cdot \sigma_{KN}}{\sigma_T} $$

Simplificando matemáticamente la contribución Compton ($\sigma_{KN}$ se cancela):

$$ \bar{E}_{tr} = \frac{E_0}{\sigma_T} \left( \sigma_{ph} + Z \cdot \sigma_{tr} \right) $$

Sustituyendo los valores numéricos:
$$ \bar{E}_{tr} = \frac{50\text{ keV}}{68.0 \times 10^{-24}\text{ cm}^2} \left[ 50.8 + 20 \times 0.04 \right] \times 10^{-24}\text{ cm}^2 $$
$$ \bar{E}_{tr} = \frac{50\text{ keV}}{68.0} \left[ 50.8 + 0.8 \right] = \frac{50 \times 51.6}{68.0}\text{ keV} \approx \mathbf{37.9\text{ keV}} $$

---

#### **3. Cálculo de la Energía Media Absorbida por Colisión ($\bar{E}_{ab}$)**
La energía absorbida en el medio ($\bar{E}_{ab}$) es la energía transferida a los electrones que finalmente se disipa en forma de ionización y excitación local, restándole las pérdidas por radiación secundaria de los propios electrones (radiación de frenado o *Bremsstrahlung*):

$$ \bar{E}_{ab} = \bar{E}_{tr} (1 - g) $$

donde $g$ es la fracción de la energía de los electrones cinéticos que se pierde por Bremsstrahlung al desacelerarse en el medio.

Debido a que la energía de los electrones puestos en movimiento es sumamente baja (menor que la energía del haz de fotones incidentes, $E_e < 50\text{ keV}$), la probabilidad de emitir radiación de frenado en un material con número atómico moderado como el Calcio es prácticamente nula ($g \approx 0$).

Por lo tanto, no existen pérdidas radiativas secundarias apreciables y la energía absorbida es exactamente igual a la energía transferida:

$$ \bar{E}_{ab} = \bar{E}_{tr} \approx \mathbf{37.9\text{ keV}} $$

```python
# 1. DATOS DE ENTRADA
E0_keV = 50.0      # Energía de incidencia (keV)
Z = 20             # Número atómico del Calcio (Ca)

# Secciones eficaces en barns (10^-24 cm^2)
s_coh = 6.0        # Coherente (atómica)
s_ph = 50.8        # Fotoeléctrico (atómica)

# Datos Compton (por electrón libre)
s_sc_elec = 0.52   # Scattering Compton por electrón
s_tr_elec = 0.04   # Transferencia Compton por electrón

# =====================================================================
# CÁLCULOS FÍSICOS
# =====================================================================
# Sección eficaz Compton total por electrón (sigma_KN = sigma_sc + sigma_tr)
s_KN_elec = s_sc_elec + s_tr_elec

# Sección eficaz incoherente por átomo (Z * sigma_KN)
s_inc_atom = Z * s_KN_elec

# Sección eficaz total por átomo
s_total = s_coh + s_ph + s_inc_atom

# Energía media transferida (E_tr)
# E_tr = (E0 / s_total) * (s_ph + Z * s_tr_elec)
E_tr_keV = (E0_keV / s_total) * (s_ph + Z * s_tr_elec)

# Energía media absorbida (E_ab)
# Para electrones de baja energía (< 50 keV), la fracción de pérdidas
# por radiación de frenado (g) es despreciable (g = 0)
g_pérdidas_radiativas = 0.0
E_ab_keV = E_tr_keV * (1.0 - g_pérdidas_radiativas)

# =====================================================================
# SALIDA DE RESULTADOS
# =====================================================================
print("=== RESOLUCIÓN NUMÉRICA AUTOMATIZADA (PROBLEMA 6) ===")
print(f"Sección eficaz incoherente (átomo): {s_inc_atom:6.2f} barns")
print(f"Sección eficaz total (átomo):       {s_total:6.2f} barns")
print("-" * 50)
print(f"Energía media transferida (E_tr):   {E_tr_keV:6.3f} keV")
print(f"Energía media absorbida (E_ab):      {E_ab_keV:6.3f} keV")
print("-" * 50)
print("Física del proceso: Dado que la energía es baja (50 keV), la fracción")
print("de pérdidas de energía por radiación de frenado (g) es estrictamente 0,")
print("por lo que la energía transferida y absorbida coinciden perfectamente.")
```

### Problema 7

La sección eficaz fotoeléctrica para la interacción de rayos $\gamma$ de $1.0\text{ MeV}$ con plomo es de $\sigma_{ph} = 6.03 \times 10^{-28}\text{ m}^2/\text{átomo}$ y la de scattering coherente vale $\sigma_{coh} = 0.911 \times 10^{-28}\text{ m}^2/\text{átomo}$. Las secciones eficaces de scattering y transferencia de energía para el efecto Compton a $1.0\text{ MeV}$ valen $\sigma_{sc} = 0.118 \times 10^{-28}\text{ m}^2/\text{electrón}$ y $\sigma_{tr} = 0.093 \times 10^{-28}\text{ m}^2/\text{electrón}$ respectivamente. Calcule:

* **a)** El **coeficiente de atenuación másico** para scattering incoherente ($\mu_{inc}/\rho$).
* **b)** El **coeficiente másico total** ($\mu/\rho$) y el **de transferencia de energía** ($\mu_{tr}/\rho$).
* **c)** El **espesor de plomo** necesario para reducir en un $5\%$ la intensidad de un haz de fotones de $1\text{ MeV}$.
* **d)** La **energía macroscópica total (en Julios)** transferida a los electrones cuando un total de $10^{15}$ fotones de $1\text{ MeV}$ atraviesan la lámina de plomo del apartado anterior.

**Datos:**
* Densidad del plomo: $\rho = 11.3 \times 10^3\text{ kg}\cdot\text{m}^{-3}$ (equivalente a $11.3\text{ g}\cdot\text{cm}^{-3}$).
* Número de electrones por kilogramo de plomo: $N_{e} = 2.38 \times 10^{26}\text{ electrones/kg}$.
* Número de átomos por kilogramo de plomo: $N_{at} = 2.91 \times 10^{24}\text{ átomos/kg}$.
* Conversión de energía: $1\text{ eV} = 1.602 \times 10^{-19}\text{ J}$.

Este ejercicio nos permite analizar de forma cuantitativa la atenuación y el depósito de energía de un haz de radiación de energía moderada ($1.0\text{ MeV}$) al atravesar un blindaje pesado como el Plomo.

---

#### **a) Coeficiente de atenuación másico para scattering incoherente ($\mu_{inc}/\rho$)**
A diferencia de los procesos fotoeléctrico y coherente (que se dan con el átomo completo), las secciones eficaces del efecto Compton se expresan **por electrón libre**.

Para hallar el coeficiente másico (que mide la probabilidad de interacción por unidad de masa superficial, en $\text{m}^2/\text{kg}$), debemos multiplicar la sección eficaz Compton total por electrón ($\sigma_{KN} = \sigma_{sc} + \sigma_{tr}$) por el número de electrones que contiene un kilogramo de plomo ($N_e$):

$$ \sigma_{KN} = \sigma_{sc} + \sigma_{tr} = (0.118 + 0.093) \times 10^{-28}\text{ m}^2/\text{e}^- = 0.211 \times 10^{-28}\text{ m}^2/\text{e}^- $$

$$ \frac{\mu_{inc}}{\rho} = N_e \cdot \sigma_{KN} $$
$$ \frac{\mu_{inc}}{\rho} = \left(2.38 \times 10^{26}\text{ e}^-/\text{kg}\right) \times \left(0.211 \times 10^{-28}\text{ m}^2/\text{e}^-\right) = 0.50218 \times 10^{-2}\text{ m}^2/\text{kg} $$

$$ \frac{\mu_{inc}}{\rho} \approx \mathbf{5.02 \times 10^{-3}\text{ m}^2/\text{kg}} $$

---

#### **b) Coeficiente másico total ($\mu/\rho$) y de transferencia ($\mu_{tr}/\rho$)**

##### **1. Coeficiente másico total ($\mu/\rho$):**
Primero calculamos las contribuciones atómicas fotoeléctrica ($\mu_{ph}/\rho$) y coherente ($\mu_{coh}/\rho$) multiplicando sus secciones eficaces por el número de átomos de plomo por kilogramo ($N_{at}$):

*   **Efecto Fotoeléctrico:**
    $$ \frac{\mu_{ph}}{\rho} = N_{at} \cdot \sigma_{ph} = \left(2.91 \times 10^{24}\text{ át/kg}\right) \times \left(6.03 \times 10^{-28}\text{ m}^2/\text{át}\right) \approx 1.75 \times 10^{-3}\text{ m}^2/\text{kg} $$
*   **Scattering Coherente:**
    $$ \frac{\mu_{coh}}{\rho} = N_{at} \cdot \sigma_{coh} = \left(2.91 \times 10^{24}\text{ át/kg}\right) \times \left(0.911 \times 10^{-28}\text{ m}^2/\text{át}\right) \approx 0.26 \times 10^{-3}\text{ m}^2/\text{kg} $$

Sumando todas las contribuciones obtenemos el coeficiente de atenuación másico total:
$$ \frac{\mu}{\rho} = \frac{\mu_{ph}}{\rho} + \frac{\mu_{coh}}{\rho} + \frac{\mu_{inc}}{\rho} $$
$$ \frac{\mu}{\rho} = (1.75 + 0.26 + 5.02) \times 10^{-3}\text{ m}^2/\text{kg} \approx \mathbf{7.03 \times 10^{-3}\text{ m}^2/\text{kg}} $$

##### **2. Coeficiente másico de transferencia de energía ($\mu_{tr}/\rho$):**
Analizamos la transferencia de energía de cada componente:
*   El **scattering coherente** es elástico, no cede energía al medio local: $\mu_{tr}^{coh}/\rho = 0$.
*   Para el **efecto fotoeléctrico**, despreciando la fluorescencia secundaria en esta aproximación, se asume transferencia total: $\mu_{tr}^{ph}/\rho = \mu_{ph}/\rho$.
*   Para el **efecto Compton**, la transferencia de energía viene determinada por su sección eficaz de transferencia específica ($\sigma_{tr}$): $$ \frac{\mu_{tr}^{inc}}{\rho} = N_e \cdot \sigma_{tr} = \left(2.38 \times 10^{26}\right) \times \left(0.093 \times 10^{-28}\right) \approx 2.21 \times 10^{-3}\text{ m}^2/\text{kg} $$

El coeficiente de transferencia de energía total del medio es:
$$ \frac{\mu_{tr}}{\rho} = \frac{\mu_{ph}}{\rho} + \frac{\mu_{tr}^{inc}}{\rho} $$
$$ \frac{\mu_{tr}}{\rho} = (1.75 + 2.21) \times 10^{-3}\text{ m}^2/\text{kg} \approx \mathbf{3.96 \times 10^{-3}\text{ m}^2/\text{kg}} $$

---

#### **c) Espesor de plomo para reducir la intensidad en un 5%**
Una reducción de intensidad del $5\%$ significa que la intensidad del haz transmitido es el $95\%$ de la incidente ($I/I_0 = 0.95$). Aplicando la ley exponencial de atenuación:

$$ \frac{I}{I_0} = e^{-\left(\frac{\mu}{\rho}\right) \cdot \rho x} = 0.95 $$

Despejando el espesor de masa superficial ($\rho x$):
$$ \rho x = -\frac{\ln(0.95)}{\mu/\rho} = -\frac{-0.05129}{7.03 \times 10^{-3}\text{ m}^2/\text{kg}} \approx 7.30\text{ kg/m}^2 $$

Dividiendo por la densidad lineal del plomo ($\rho = 11.3 \times 10^3\text{ kg/m}^3$), hallamos el espesor físico $x$:
$$ x = \frac{\rho x}{\rho} = \frac{7.30\text{ kg/m}^2}{11.3 \times 10^3\text{ kg/m}^3} \approx 6.46 \times 10^{-4}\text{ m} \approx \mathbf{0.646\text{ mm}} $$

---

#### **d) Energía macroscópica transferida a los electrones ($E_{transf}$)**

1. **Energía media transferida por colisión ($\bar{E}_{tr}$):**
   De forma estadística, cada fotón de $1.0\text{ MeV}$ que experimenta una colisión en el plomo transfiere a los electrones del medio una energía media dada por la relación de coeficientes:
   $$ \bar{E}_{tr} = E_0 \cdot \frac{\mu_{tr}/\rho}{\mu/\rho} = 1.0\text{ MeV} \times \frac{3.96 \times 10^{-3}}{7.03 \times 10^{-3}} \approx 0.563\text{ MeV} $$

2. **Número de fotones que interaccionan ($N_{int}$):**
   Como el haz se atenúa un $5\%$, la probabilidad de interacción de cada fotón en la lámina es exactamente $P = 0.05$. Para un total de $N = 10^{15}$ fotones incidentes:
   $$ N_{int} = N \cdot P = 10^{15} \times 0.05 = 5.0 \times 10^{13}\text{ fotones} $$

3. **Cálculo de la Energía Total en Julios:**
   La energía cinética transferida a los electrones del blindaje es el producto de los fotones que colisionan por la energía media depositada en cada colisión:
   $$ E_{transf} = N_{int} \cdot \bar{E}_{tr} = \left(5.0 \times 10^{13}\text{ fotones}\right) \times \left(0.563\text{ MeV/fotón}\right) = 2.815 \times 10^{13}\text{ MeV} $$

   Convertimos esta energía a Julios ($1\text{ MeV} = 1.602 \times 10^{-13}\text{ J}$):
   $$ E_{transf} = \left(2.815 \times 10^{13}\text{ MeV}\right) \times \left(1.60 \times 10^{-13}\text{ J/MeV}\right) \approx \mathbf{4.51\text{ Julios}} $$

```python
# 1. DATOS DE ENTRADA (ENUNCIADO)
E0_MeV = 1.0        # Energía de incidencia (MeV)
N_gamma = 1e15      # Número de fotones incidentes
rho = 11.3e3        # Densidad del plomo (kg/m^3)
N_e = 2.38e26       # Electrones por kg de Pb
N_at = 2.91e24      # Átomos por kg de Pb

# Secciones eficaces en m^2 (enunciado tiene factor 10^-28)
factor_seccion = 1e-28
s_ph = 6.03 * factor_seccion
s_coh = 0.911 * factor_seccion
s_sc = 0.118 * factor_seccion
s_tr = 0.093 * factor_seccion

# Constantes de conversión
MeV_to_J = 1.602176634e-13  # Conversión exacta de MeV a Julios
MeV_to_J_approx = 1.60e-13  # Conversión aproximada habitual

# =====================================================================
# DESARROLLO MATEMÁTICO
# =====================================================================

# --- APARTADO A ---
s_inc_elec = s_sc + s_tr
mu_inc_over_rho = N_e * s_inc_elec

# --- APARTADO B ---
# Atenuaciones individuales
mu_ph_over_rho = N_at * s_ph
mu_coh_over_rho = N_at * s_coh

# Coeficiente másico total (mu/rho)
mu_over_rho_analytic = mu_ph_over_rho + mu_coh_over_rho + mu_inc_over_rho

# Coeficiente de transferencia (mu_tr/rho)
mu_tr_inc_over_rho = N_e * s_tr
mu_tr_over_rho_analytic = mu_ph_over_rho + mu_tr_inc_over_rho

# --- APARTADO C ---
Atenuacion_pct = 0.05
I_I0 = 1.0 - Atenuacion_pct  # 0.95

# Espesor de masa y lineal analíticos puros
rho_x_analytic = -np.log(I_I0) / mu_over_rho_analytic
x_m_analytic = rho_x_analytic / rho

# --- APARTADO D ---
# Probabilidad de interacción en este espesor exacto
P_int = Atenuacion_pct  # 5%
N_int = N_gamma * P_int

# Energía transferida por colisión
E_tr_col_analytic = E0_MeV * (mu_tr_over_rho_analytic / mu_over_rho_analytic)

# Energía macroscópica total transferida
E_total_J_analytic = N_int * E_tr_col_analytic * MeV_to_J

# =====================================================================
# IMPRESIÓN DE RESULTADOS COMPARATIVOS
# =====================================================================
print("================================================================")
print("             RESULTADOS NUMÉRICOS (PROBLEMA 7)                  ")
print("================================================================")
print(f"a) Coeficiente másico incoherente (μ_inc/ρ):")
print(f"   -> {mu_inc_over_rho:.5e} m²/kg (Aprox: 5.02e-3 m²/kg)")
print("-" * 64)
print(f"b) Coeficiente másico total (μ/ρ):")
print(f"   -> Resultado: {mu_over_rho_analytic:.5e} m²/kg")
print(f"   Coeficiente másico de transferencia (μ_tr/ρ):")
print(f"   -> Resultado: {mu_tr_over_rho_analytic:.5e} m²/kg")
print("-" * 64)
print(f"c) Espesor lineal para atenuar un 5%:")
print(f"   -> Resultado: {x_m_analytic*1e3:.4f} mm ({x_m_analytic:.4e} m)")
print("-" * 64)
print(f"d) Energía macroscópica total transferida:")
print(f"   -> Energía media/colisión: {E_tr_col_analytic:.4f} MeV")
print(f"   -> Resultado:         {E_total_J_analytic:.4f} J")
print("================================================================")

sns.set_theme(style='whitegrid', palette='colorblind', font='DejaVu Sans')

# ---------------------------------------------------------
# PARÁMETROS FÍSICOS (Plomo a 1.0 MeV)
# ---------------------------------------------------------
mu_rho = mu_over_rho_analytic # Coeficiente de atenuación másico (m^2/kg)
mu_m = mu_rho * rho           # Coeficiente de atenuación lineal (m^-1)
mu_mm = mu_m / 1000           # Convertido a mm^-1 para el eje X
x_max = 20
# Rango de espesores a simular (de 0 a 5 milímetros)
espesores = np.linspace(0, x_max, 1000)

# Ley exponencial de atenuación: I/I0 = e^(-mu * x)
transmision_pct = np.exp(-mu_mm * espesores) * 100
atenuacion_pct = (1.0 - np.exp(-mu_mm * espesores)) * 100

# Punto exacto de reducción del 5% obtenido en el apartado c)
x_target = 0.646
y_trans_target = 95.0
y_aten_target = 5.0

# ---------------------------------------------------------
# GENERACIÓN DE LA GRÁFICA
# ---------------------------------------------------------
fig, ax = plt.subplots(figsize=(10, 6))

# Dibujar las curvas de transmisión y de atenuación
linea_trans, = ax.plot(espesores, transmision_pct, color='#4c72b0', linewidth=2.5,
                       label='Intensidad Transmitida ($I/I_0$)')
linea_aten, = ax.plot(espesores, atenuacion_pct, color='#c44e52', linewidth=2.5, linestyle='--',
                      label='Fracción Atenuada ($1 - I/I_0$)')

# Resaltar el punto de trabajo del ejercicio con marcadores específicos
ax.plot(x_target, y_trans_target, 'o', color='darkblue', markersize=8)
ax.plot(x_target, y_aten_target, 's', color='darkred', markersize=8)

# Añadir líneas guía punteadas para conectar los ejes con el punto de interés
ax.axvline(x=x_target, color='gray', linestyle=':', alpha=0.7)
ax.axhline(y=y_trans_target, color='gray', linestyle=':', alpha=0.7)
ax.axhline(y=y_aten_target, color='gray', linestyle=':', alpha=0.7)

# Etiqueta flotante con flecha indicando los valores exactos
ax.annotate(f'Espesor: {x_target:.3f} mm\nTransmisión: {y_trans_target:.1f}%\nAtenuación: {y_aten_target:.1f}%',
             xy=(x_target, y_trans_target),
             xytext=(x_target + 3, y_trans_target - 15),
             arrowprops=dict(arrowstyle="->", color='black', lw=1.2),
             fontsize=10.5,
             bbox=dict(boxstyle="round,pad=0.4", facecolor='white', edgecolor='gray', alpha=0.9))

# Configuración de límites y formatos de ejes
ax.set_xlim(0, x_max)
ax.set_ylim(-3, 103)
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda y, _: f'{int(y)}%'))

# Título descriptivo enfocado en la física y etiquetas de magnitud
ax.set_title('Curva de atenuación y transmisión exponencial en Plomo a 1.0 MeV',
             fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel('Espesor de Plomo (mm)', fontsize=12)
ax.set_ylabel('Porcentaje del haz de fotones (%)', fontsize=12)

# Posicionamiento de la leyenda
ax.legend(handles=[linea_trans, linea_aten], loc='center right', fontsize=11)

# Eliminar líneas de contorno superior y derecha para una estética limpia
sns.despine()

# Ajustar márgenes para evitar recortes de texto
plt.tight_layout()
plt.show()
```

### Problema 8

La interacción de fotones de $100\text{ MeV}$ con átomos de Hidrógeno ($H$, $Z = 1$, $P_a = 1.01\text{ g/mol}$) y Oxígeno ($O$, $Z = 8$, $P_a = 16.00\text{ g/mol}$) tiene lugar básicamente por scattering incoherente o creación de pares.

Sabiendo que las secciones eficaces Compton (por electrón) de scattering y transferencia valen $\sigma_{sc}^c = 1.7 \times 10^{-27}\text{ cm}^2$ y $\sigma_{tr}^c = 6.5 \times 10^{-27}\text{ cm}^2$, y que las secciones eficaces de creación de pares valen $\sigma_{par}^H = 1.16 \times 10^{-26}\text{ cm}^2$ y $\sigma_{par}^O = 4.10 \times 10^{-25}\text{ cm}^2$, calcule:

* **a)** La **energía media transferida** por colisión para el átomo de Hidrógeno ($\bar{E}_{tr}^H$) y de Oxígeno ($\bar{E}_{tr}^O$).
* **b)** La **energía media transferida** por colisión para la molécula de agua ($\bar{E}_{tr}^{H_2O}$).
* **c)** El **coeficiente de atenuación másico** del agua ($\mu/\rho$).

**Dato:** Densidad del agua: $\rho = 1.00\text{ g}\cdot\text{cm}^{-3}$.


En el rango de energías de los rayos $\gamma$ de alta energía ($E_0 = 100\text{ MeV}$), los efectos fotoeléctrico y coherente son despreciables frente a la creación de pares y el scattering incoherente (Compton).

---

#### **1. Secciones eficaces de colisión e incoherente**

Las secciones eficaces Compton se proporcionan **por electrón libre** ($\sigma^c$). Para obtener las secciones eficaces atómicas de scattering incoherente, multiplicamos por el número atómico $Z$ del elemento correspondiente:

*   **Para el Hidrógeno ($Z = 1$):**
    $$ \sigma_{inc}^H = Z_H \cdot (\sigma_{sc}^c + \sigma_{tr}^c) = 1 \cdot (1.7 + 6.5) \times 10^{-27}\text{ cm}^2 = 8.2 \times 10^{-27}\text{ cm}^2 = 0.082 \times 10^{-25}\text{ cm}^2 $$
*   **Para el Oxígeno ($Z = 8$):**
    $$ \sigma_{inc}^O = Z_O \cdot (\sigma_{sc}^c + \sigma_{tr}^c) = 8 \cdot (1.7 + 6.5) \times 10^{-27}\text{ cm}^2 = 65.6 \times 10^{-27}\text{ cm}^2 = 0.656 \times 10^{-25}\text{ cm}^2 $$

---

#### **2. Secciones eficaces atómicas totales ($\sigma_T$)**

Sumamos las contribuciones del scattering incoherente y la creación de pares para cada átomo:

*   **Hidrógeno (con $\sigma_{par}^H = 1.16 \times 10^{-26}\text{ cm}^2 = 0.116 \times 10^{-25}\text{ cm}^2$):**
    $$ \sigma_T^H = \sigma_{inc}^H + \sigma_{par}^H = (0.082 + 0.116) \times 10^{-25}\text{ cm}^2 = \mathbf{0.198 \times 10^{-25}\text{ cm}^2} \quad (0.00198\text{ barns}) $$
*   **Oxígeno (con $\sigma_{par}^O = 4.10 \times 10^{-25}\text{ cm}^2$):**
    $$ \sigma_T^O = \sigma_{inc}^O + \sigma_{par}^O = (0.656 + 4.10) \times 10^{-25}\text{ cm}^2 = \mathbf{4.756 \times 10^{-25}\text{ cm}^2} \quad (0.04756\text{ barns}) $$

---

#### **3. Energía media transferida por proceso**

*   **Scattering Incoherente (Compton):**
    La fracción de energía transferida al electrón de retroceso viene dada por el cociente entre la sección eficaz de transferencia de energía Compton y la sección eficaz Compton total:
    
  $$ \bar{E}_{tr}^{inc} = E_0 \cdot \frac{\sigma_{tr}^c}{\sigma_{tr}^c + \sigma_{sc}^c} = 100\text{ MeV} \times \frac{6.5 \times 10^{-27}}{8.2 \times 10^{-27}} \approx 79.27\text{ MeV} $$

*   **Creación de Pares:**
    El fotón desaparece y transfiere toda su energía al par electrón-positrón excepto la energía umbral requerida para crear sus masas en reposo ($2 m_e c^2 \approx 1.022\text{ MeV}$):
    $$ \bar{E}_{tr}^{par} = E_0 - 2m_e c^2 = 100\text{ MeV} - 1.022\text{ MeV} = 98.98\text{ MeV} $$

---

#### **a) Energía media transferida por colisión para H y O**

Ponderamos las energías transferidas en cada proceso por la probabilidad de que ocurra dicho proceso para cada átomo:

*   **Para el Hidrógeno (H):**
   $$ \bar{E}_{tr}^H = \bar{E}_{tr}^{inc} \frac{\sigma_{inc}^H}{\sigma_T^H} + \bar{E}_{tr}^{par} \frac{\sigma_{par}^H}{\sigma_T^H} $$

   $$ \bar{E}_{tr}^H = 79.27\text{ MeV} \times \frac{0.082}{0.198} + 98.98\text{ MeV} \times \frac{0.116}{0.198} \approx 32.83 + 57.99 = \mathbf{90.8\text{ MeV}} $$

*   **Para el Oxígeno (O):**
   $$ \bar{E}_{tr}^O = \bar{E}_{tr}^{inc} \frac{\sigma_{inc}^O}{\sigma_T^O} + \bar{E}_{tr}^{par} \frac{\sigma_{par}^O}{\sigma_T^O} $$

   $$ \bar{E}_{tr}^O = 79.27\text{ MeV} \times \frac{0.656}{4.756} + 98.98\text{ MeV} \times \frac{4.10}{4.756} \approx 10.93 + 85.39 = \mathbf{96.3\text{ MeV}} $$

---

#### **b) Energía media transferida por colisión para el agua ($\bar{E}_{tr}^{H_2O}$)**

Una molécula de agua está compuesta por dos átomos de Hidrógeno y uno de Oxígeno. La sección eficaz total de interacción de la molécula es aditiva:
$$ \sigma_T^{H_2O} = 2 \cdot \sigma_T^H + \sigma_T^O $$
$$ \sigma_T^{H_2O} = 2 \cdot (0.198 \times 10^{-25}) + 4.756 \times 10^{-25} = 5.152 \times 10^{-25}\text{ cm}^2 $$

La energía media transferida por colisión molecular es el promedio ponderado de las energías medias de cada átomo según su contribución a la sección eficaz molecular total:

$$ \bar{E}_{tr}^{H_2O} = \frac{2 \cdot \sigma_T^H}{\sigma_T^{H_2O}} \bar{E}_{tr}^H + \frac{\sigma_T^O}{\sigma_T^{H_2O}} \bar{E}_{tr}^O $$
$$ \bar{E}_{tr}^{H_2O} = \frac{2 \cdot 0.198}{5.152} \cdot 90.8\text{ MeV} + \frac{4.756}{5.152} \cdot 96.3\text{ MeV} \approx 6.98 + 88.92 = \mathbf{95.9\text{ MeV}} $$

---

#### **c) Coeficiente de atenuación másico del agua ($\mu/\rho$)**

El coeficiente másico de atenuación del agua se calcula a partir de la sección eficaz molecular total ($\sigma_T^{H_2O}$) y su masa molecular ($P_a^{H_2O} = 2 \cdot 1.01 + 16.00 = 18.02\text{ g/mol}$):

$$ \frac{\mu}{\rho} = \frac{N_A}{P_a^{H_2O}} \cdot \sigma_T^{H_2O} $$

Sustituyendo las constantes:
$$ \frac{\mu}{\rho} = \frac{6.022 \times 10^{23}\text{ moléculas/mol}}{18.02\text{ g/mol}} \times \left(5.152 \times 10^{-25}\text{ cm}^2\right) $$
$$ \frac{\mu}{\rho} = 0.33418 \times 10^{23} \times 5.152 \times 10^{-25}\text{ cm}^2/\text{g} \approx \mathbf{0.0172\text{ cm}^2/\text{g}} $$

*(Alternativamente, se puede llegar al mismo resultado aplicando la suma de los coeficientes másicos individuales de cada átomo ponderados por sus fracciones en masa, $w_H \approx 0.112$ y $w_O \approx 0.888$, obteniéndose el mismo valor de $0.0172\text{ cm}^2/\text{g}$).*


```python
# 1. DATOS DE ENTRADA
E0_MeV = 100.0
me_c2 = 0.511

# Masas moleculares (g/mol)
M_H = 1.01
M_O = 16.00
M_H2O = 2 * M_H + M_O

# Secciones eficaces (en unidades de 10^-25 cm^2 para homogeneidad)
s_sc_c = 0.017   # 1.7 * 10^-27 cm^2
s_tr_c = 0.065   # 6.5 * 10^-27 cm^2

s_par_H = 0.116  # 1.16 * 10^-26 cm^2
s_par_O = 4.10   # 4.10 * 10^-25 cm^2

# Constantes
NA = 6.0221415e23

# =====================================================================
# CÁLCULOS
# =====================================================================

# --- Scattering Compton atómico (Z * sigma_Compton) ---
s_inc_H = 1 * (s_sc_c + s_tr_c)
s_inc_O = 8 * (s_sc_c + s_tr_c)

# --- Secciones eficaces totales atómicas ---
s_total_H = s_inc_H + s_par_H
s_total_O = s_inc_O + s_par_O

# --- Energías de transferencia medias por proceso ---
# Compton: E_tr_Compton = E0 * sigma_tr / (sigma_tr + sigma_sc)
E_tr_inc = E0_MeV * (s_tr_c / (s_tr_c + s_sc_c))

# Creación de pares: E_tr_pares = E0 - 2 * me_c2
E_tr_par = E0_MeV - 2 * me_c2

# --- Energía media transferida por átomo (H y O) ---
E_tr_H = (s_inc_H / s_total_H) * E_tr_inc + (s_par_H / s_total_H) * E_tr_par
E_tr_O = (s_inc_O / s_total_O) * E_tr_inc + (s_par_O / s_total_O) * E_tr_par

# --- Energía media transferida por molécula de agua (H2O) ---
s_total_H2O = 2 * s_total_H + s_total_O
E_tr_H2O = ((2 * s_total_H) / s_total_H2O) * E_tr_H + (s_total_O / s_total_H2O) * E_tr_O

# --- Coeficiente de atenuación másico del agua (mu/rho) ---
# s_total_H2O está en 10^-25 cm^2, multiplicamos por 10^-25 para cm^2 reales
mu_rho_H2O = (NA / M_H2O) * (s_total_H2O * 1e-25)

# --- Coeficientes másicos individuales por átomo ---
mu_rho_H = (NA / M_H) * (s_total_H * 1e-25)
mu_rho_O = (NA / M_O) * (s_total_O * 1e-25)

# Fracciones en masa
w_H = (2 * M_H) / M_H2O
w_O = M_O / M_H2O
mu_rho_H2O_frac = w_H * mu_rho_H + w_O * mu_rho_O

# =====================================================================
# IMPRESIÓN DE RESULTADOS
# =====================================================================
print("=====================================================================")
print("                            RESULTADOS                               ")
print("=====================================================================")
print(f"a) Energía media transferida por átomo:")
print(f"   -> Hidrógeno (H): {E_tr_H:.4f} MeV")
print(f"   -> Oxígeno (O):   {E_tr_O:.4f} MeV")
print("-" * 69)
print(f"b) Energía media transferida por molécula de agua:")
print(f"   -> Agua (H2O):    {E_tr_H2O:.4f} MeV")
print("-" * 69)
print(f"c) Coeficiente de atenuación másico (μ/ρ) del agua:")
print(f"   -> Por sección molecular: {mu_rho_H2O:.6f} cm²/g)")
print(f"   -> Por fracción de masa:  {mu_rho_H2O_frac:.6f} cm²/g")
print("=====================================================================")
```

### **Problema 9:**

El Germanato de Bismuto, $\text{Bi}_4\text{Ge}_3\text{O}_{12}$, también llamado BGO, es un material centelleador de alta densidad, $\rho = 7.13\text{ g}\cdot\text{cm}^{-3}$, que absorbe muy eficientemente la radiación $\gamma$. Empleando los datos de sección eficaz para creación de pares de los 3 elementos que lo componen:

| Elemento | $Z$ | Masa Atómica ($P_a$, g/mol) | $\sigma_{par}$ (barns) |
| :--- | :---: | :---: | :---: |
| **Bismuto ($\text{Bi}$)** | $83$ | $209.0$ | $33.9$ |
| **Germanio ($\text{Ge}$)** | $32$ | $72.6$ | $5.88$ |
| **Oxígeno ($\text{O}$)** | $8$ | $16.0$ | $0.40$ |

Calcule:

* **a)** El **coeficiente de atenuación másico** ($\mu/\rho$) y el **de transferencia de energía** ($\mu_{tr}/\rho$) del BGO para rayos gamma de $150\text{ MeV}$.
* **b)** El **espesor de BGO** necesario para reducir al $10\%$ la intensidad de un haz de rayos $\gamma$ de esta energía.

Para resolver este ejercicio en el rango de muy alta energía ($E_0 = 150\text{ MeV}$), consideramos que la interacción de fotones con la materia está completamente dominada por la **creación de pares**. Por tanto, la sección eficaz total de interacción atómica y molecular coincide de forma aproximada con la sección eficaz de creación de pares.

---

#### **a) Coeficientes de atenuación y de transferencia de energía**

##### **1. Masa molecular del BGO ($P_{BGO}$)**
La fórmula del Germanato de Bismuto es $\text{Bi}_4\text{Ge}_3\text{O}_{12}$. Calculamos la masa molecular sumando las contribuciones de sus átomos constituyentes:
$$ P_{BGO} = 4 \cdot P_{Bi} + 3 \cdot P_{Ge} + 12 \cdot P_{O} $$
$$ P_{BGO} = 4(209.0) + 3(72.6) + 12(16.0) = 836.0 + 217.8 + 192.0 = 1245.8\text{ g/mol} $$

##### **2. Sección eficaz molecular del BGO ($\sigma_{BGO}$)**
Al tratarse de una mezcla homogénea, la sección eficaz molecular para el proceso de creación de pares es la suma aditiva de las secciones eficaces individuales de cada átomo:
$$ \sigma_{BGO} = 4 \cdot \sigma_{par}^{Bi} + 3 \cdot \sigma_{par}^{Ge} + 12 \cdot \sigma_{par}^{O} $$
$$ \sigma_{BGO} = 4(33.9) + 3(5.88) + 12(0.40)\text{ barns} $$
$$ \sigma_{BGO} = 135.6 + 17.64 + 4.8 = 158.04\text{ barns} = 158.04 \times 10^{-24}\text{ cm}^2 $$

##### **3. Coeficiente de atenuación másico del BGO ($\mu/\rho$)**
Utilizando la sección eficaz molecular y la masa molecular de la sustancia, calculamos el coeficiente másico total de interacción:
$$ \left(\frac{\mu}{\rho}\right)_{BGO} = \frac{N_A}{P_{BGO}} \cdot \sigma_{BGO} $$

Sustituyendo los valores de la constante de Avogadro ($N_A \approx 6.022 \times 10^{23}\text{ mol}^{-1}$):
$$ \left(\frac{\mu}{\rho}\right)_{BGO} = \frac{6.022 \times 10^{23}}{1245.8\text{ g/mol}} \times \left(158.04 \times 10^{-24}\text{ cm}^2\right) \approx \mathbf{0.0764\text{ cm}^2\cdot\text{g}^{-1}} $$

##### **4. Coeficiente másico de transferencia de energía ($\mu_{tr}/\rho$)**
En el proceso de creación de pares, el fotón desaparece y toda su energía inicial ($E_0$) se transfiere en forma de energía cinética a la pareja electrón-positrón, a excepción de la energía umbral requerida para generar sus masas en reposo ($2 m_e c^2 \approx 1.022\text{ MeV}$):
$$ \bar{E}_{tr} = E_0 - 2m_e c^2 = 150 - 1.022 = 148.98\text{ MeV} $$

La fracción de energía transferida al medio por interacción es $\bar{E}_{tr}/E_0$. El coeficiente másico de transferencia de energía se calcula multiplicando el coeficiente de atenuación por esta fracción:
$$ \left(\frac{\mu_{tr}}{\rho}\right)_{BGO} = \left(\frac{\mu}{\rho}\right)_{BGO} \times \frac{\bar{E}_{tr}}{E_0} $$
$$ \left(\frac{\mu_{tr}}{\rho}\right)_{BGO} = 0.0764\text{ cm}^2/\text{g} \times \frac{148.98\text{ MeV}}{150\text{ MeV}} \approx \mathbf{0.0759\text{ cm}^2\cdot\text{g}^{-1}} $$

---

#### **b) Espesor de BGO para reducir la intensidad al 10%**

Buscamos reducir la intensidad del haz de fotones incidentes al $10\%$ ($I/I_0 = 0.10$). Utilizando la ley de atenuación exponencial:
$$ \frac{I}{I_0} = e^{-\left(\frac{\mu}{\rho}\right) \cdot \rho x} = 0.10 $$

Aplicando logaritmo natural y despejando el espesor de masa superficial ($\rho x$):
$$ \rho x = -\frac{\ln(0.10)}{\left(\frac{\mu}{\rho}\right)_{BGO}} \approx \frac{2.3026}{0.0764\text{ cm}^2/\text{g}} \approx 30.14\text{ g}\cdot\text{cm}^{-2} $$

Para obtener el espesor lineal físico ($x$), dividimos el espesor de masa por la densidad volumétrica del cristal de BGO ($\rho = 7.13\text{ g}\cdot\text{cm}^{-3}$):
$$ x = \frac{\rho x}{\rho} = \frac{30.14\text{ g/cm}^2}{7.13\text{ g/cm}^3} \approx \mathbf{4.22\text{ cm}}$$

```python
# 1. DATOS DE ENTRADA
E0_MeV = 150.0
me_c2 = 0.511
rho_BGO = 7.13    # Densidad en g/cm^3

# Datos atómicos de los componentes
# Elemento: [Número de átomos, Masa atómica (g/mol), sigma_par (barns)]
composicion = {
    'Bi': [4, 209.0, 33.9],
    'Ge': [3, 72.6, 5.88],
    'O':  [12, 16.0, 0.40]
}

NA = 6.0221415e23  # Número de Avogadro

# =====================================================================
# CÁLCULOS
# =====================================================================

# a) Masa molecular del BGO (g/mol)
M_BGO = sum(datos[0] * datos[1] for datos in composicion.values())

# b) Sección eficaz molecular total de creación de pares (barns)
# 1 barn = 10^-24 cm^2
s_BGO_barns = sum(datos[0] * datos[2] for datos in composicion.values())
s_BGO_cm2 = s_BGO_barns * 1e-24

# c) Coeficiente másico de atenuación (mu/rho)
mu_rho = (NA / M_BGO) * s_BGO_cm2

# d) Energía media transferida (E_tr) y coeficiente de transferencia (mu_tr/rho)
E_tr_MeV = E0_MeV - 2 * me_c2
mu_tr_rho = mu_rho * (E_tr_MeV / E0_MeV)

# e) Espesor para reducir la intensidad al 10%
fraccion_I = 0.10
rho_x = -np.log(fraccion_I) / mu_rho
x_cm = rho_x / rho_BGO

# =====================================================================
# IMPRESIÓN DE RESULTADOS
# =====================================================================
print("=====================================================================")
print("                RESULTADOS NUMÉRICOS (PROBLEMA 9)                    ")
print("=====================================================================")
print(f"Masa Molecular del BGO (Bi4Ge3O12):     {M_BGO:.2f} g/mol")
print(f"Sección Eficaz Molecular Total:         {s_BGO_barns:.2f} barns")
print("-" * 69)
print(f"a) Coeficiente másico de atenuación:    {mu_rho:.5f} cm²/g (Aprox: 0.0764 cm²/g)")
print(f"   Coeficiente másico de transferencia:  {mu_tr_rho:.5f} cm²/g (Aprox: 0.0759 cm²/g)")
print("-" * 69)
print(f"b) Espesor requerido para reducir la intensidad al 10%:")
print(f"   -> Espesor másico superficial (ρx):  {rho_x:.3f} g/cm²")
print(f"   -> Espesor físico lineal (x):        {x_cm:.3f} cm       (Aprox: 4.2 cm)")
print("=====================================================================")
```

### Problema 10:

**a)** Un haz de rayos X de $50\text{ keV}$ incide sobre Calcio ($Z=20$, $A=40$). La sección eficaz de scattering coherente vale $\sigma_{coh} = 6.0 \times 10^{-24}\text{ cm}^2$ y la de efecto fotoeléctrico es $\sigma_{ph} = 50.8 \times 10^{-24}\text{ cm}^2$. Suponiendo que el scattering incoherente pueda ser descrito de forma aproximada por el efecto Compton para el que, a $50\text{ keV}$, las secciones eficaces por electrón de scattering y de transferencia valen $\sigma_{sc} = 0.52 \times 10^{-24}\text{ cm}^2$ y $\sigma_{tr} = 0.04 \times 10^{-24}\text{ cm}^2$, calcule la **energía media transferida** ($\bar{E}_{tr}$) y la **energía media absorbida** ($\bar{E}_{ab}$) por colisión.

**b)** El contraste radiológico está relacionado con el cociente de coeficientes de atenuación total entre los diferentes medios. Empleando los datos de la tabla adjunta y suponiendo que las secciones eficaces de scattering incoherente se pueden calcular a partir del efecto Compton, calcule los cocientes:

$$ \frac{(\mu/\rho)_{\text{hueso}}}{(\mu/\rho)_{\text{músculo}}} \quad \text{y} \quad \frac{\mu_{\text{hueso}}}{\mu_{\text{músculo}}} $$

para rayos X de $50\text{ keV}$, despreciando la contribución de los *Otros* elementos cuyos datos no se consignan en la tabla. Halle también el coeficiente másico de transferencia de energía $\mu_{tr}/\rho$ para músculo y hueso.

**DATOS:**
* Densidades volumétricas: $\rho_{\text{músculo}} = 1.04\text{ g}\cdot\text{cm}^{-3}$ y $\rho_{\text{hueso}} = 1.65\text{ g}\cdot\text{cm}^{-3}$.
* Composición en peso y secciones eficaces a $50\text{ keV}$:

| Elemento | $Z/A$ | % en peso Músculo | % en peso Hueso | $\sigma_{coh}$ ($10^{-24}\text{ cm}^2$) | $\sigma_{ph}$ ($10^{-24}\text{ cm}^2$) |
| :--- | :---: | :---: | :---: | :---: | :---: |
| **Hidrógeno (H)** | $1/1$ | $10\%$ | $6\%$ | — | — |
| **Carbono (C)** | $6/12$ | $13\%$ | $29\%$ | — | $0.17$ |
| **Oxígeno (O)** | $8/16$ | $73\%$ | $41\%$ | — | $0.73$ |
| **Calcio (Ca)** | $20/40$ | — | $15\%$ | $6.0$ | $50.8$ |
| **Otros** | — | $4\%$ | $9\%$ | — | — |

---

#### **a) Energía media transferida y absorbida por colisión en Calcio**

Este apartado consiste en un análisis directo del Calcio ($Z=20$, $A=40$) a la energía de $50\text{ keV}$:

1. **Sección eficaz Compton total por electrón:**
   $$ \sigma_{KN} = \sigma_{sc} + \sigma_{tr} = (0.52 + 0.04) \times 10^{-24}\text{ cm}^2 = 0.56\text{ barns/electrón} $$

2. **Sección eficaz incoherente (Compton) por átomo:**
   Multiplicamos la contribución por electrón por el número atómico del Calcio ($Z=20$):
   $$ \sigma_{inc} \approx Z \cdot \sigma_{KN} = 20 \times 0.56 \times 10^{-24}\text{ cm}^2 = 11.2\text{ barns/átomo} $$

3. **Sección eficaz total por átomo ($\sigma_T$):**
   $$ \sigma_T = \sigma_{coh} + \sigma_{ph} + \sigma_{inc} = 6.0 + 50.8 + 11.2 = 68.0\text{ barns} $$

4. **Energía media transferida ($\bar{E}_{tr}$):**
   Ponderamos las energías transferidas en el fotoeléctrico ($\bar{E}_{tr}^{ph} \approx E_0$) y Compton ($\bar{E}_{tr}^{inc} = E_0 \frac{\sigma_{tr}}{\sigma_{KN}}$) por sus probabilidades relativas:
   $$ \bar{E}_{tr} = \frac{E_0}{\sigma_T} \left( \sigma_{ph} + Z \cdot \sigma_{tr} \right) $$
   $$ \bar{E}_{tr} = \frac{50\text{ keV}}{68.0} \left( 50.8 + 20 \times 0.04 \right) = \frac{50 \times 51.6}{68.0} \approx \mathbf{37.94\text{ keV}} $$

5. **Energía media absorbida ($\bar{E}_{ab}$):**
   Al ser electrones secundarios de muy baja energía ($< 50\text{ keV}$), las pérdidas por radiación de frenado (*Bremsstrahlung*) son nulas ($g \approx 0$). Por lo tanto:
   $$ \bar{E}_{ab} = \bar{E}_{tr} (1 - g) \approx \bar{E}_{tr} \approx \mathbf{37.94\text{ keV}} $$

---

#### **b) Análisis del contraste radiológico entre músculo y hueso**

Para calcular los coeficientes de mezclas complejas como tejidos biológicos, aplicamos la ley de aditividad por fracciones en peso ($w_i$), despreciando la contribución de los elementos "Otros" tal como indica la consigna:

$$ \frac{\mu}{\rho} = \sum_{i} w_i \left(\frac{\mu}{\rho}\right)_i = \sum_{i} w_i \left( \frac{N_A}{A_i} \cdot \sigma_i \right) $$

##### **1. Coeficientes másicos individuales ($\mu/\rho$) y de transferencia ($\mu_{tr}/\rho$) a $50\text{ keV}$**

*   **Hidrógeno (H, $Z=1$, $A=1$):**
    *   $\sigma_T = \sigma_{inc} = 1 \times 0.56 = 0.56\text{ barns}$.
    *   $\sigma_{tr} = Z \times \sigma_{tr}^e = 1 \times 0.04 = 0.04\text{ barns}$.
    *   $(\mu/\rho)_H = 0.56 \times \frac{0.6022}{1.0} \approx 0.3372\text{ cm}^2/\text{g}$.
    *   $(\mu_{tr}/\rho)_H = 0.04 \times \frac{0.6022}{1.0} \approx 0.0241\text{ cm}^2/\text{g}$.

*   **Carbono (C, $Z=6$, $A=12$):**
    *   $\sigma_T = \sigma_{ph} + \sigma_{inc} = 0.17 + 6 \times 0.56 = 3.53\text{ barns}$.
    *   $\sigma_{tr} = \sigma_{ph} + Z \times \sigma_{tr}^e = 0.17 + 6 \times 0.04 = 0.41\text{ barns}$.
    *   $(\mu/\rho)_C = 3.53 \times \frac{0.6022}{12.0} \approx 0.1772\text{ cm}^2/\text{g}$.
    *   $(\mu_{tr}/\rho)_C = 0.41 \times \frac{0.6022}{12.0} \approx 0.0206\text{ cm}^2/\text{g}$.

*   **Oxígeno (O, $Z=8$, $A=16$):**
    *   $\sigma_T = \sigma_{ph} + \sigma_{inc} = 0.73 + 8 \times 0.56 = 5.21\text{ barns}$.
    *   $\sigma_{tr} = \sigma_{ph} + Z \times \sigma_{tr}^e = 0.73 + 8 \times 0.04 = 1.05\text{ barns}$.
    *   $(\mu/\rho)_O = 5.21 \times \frac{0.6022}{16.0} \approx 0.1961\text{ cm}^2/\text{g}$.
    *   $(\mu_{tr}/\rho)_O = 1.05 \times \frac{0.6022}{16.0} \approx 0.0395\text{ cm}^2/\text{g}$.

*   **Calcio (Ca, $Z=20$, $A=40$):**
    *   $\sigma_T = 68.0\text{ barns}$ (calculado en el apartado a).
    *   $\sigma_{tr} = \sigma_{ph} + Z \times \sigma_{tr}^e = 50.8 + 20 \times 0.04 = 51.6\text{ barns}$.
    *   $(\mu/\rho)_{Ca} = 68.0 \times \frac{0.6022}{40.0} \approx 1.0238\text{ cm}^2/\text{g}$.
    *   $(\mu_{tr}/\rho)_{Ca} = 51.6 \times \frac{0.6022}{40.0} \approx 0.7769\text{ cm}^2/\text{g}$.

##### **2. Coeficientes del Músculo ($w_H=0.10, w_C=0.13, w_O=0.73$):**
$$ \left(\frac{\mu}{\rho}\right)_{\text{mús}} = 0.10(0.3372) + 0.13(0.1772) + 0.73(0.1961) \approx \mathbf{0.1999\text{ cm}^2/\text{g}} $$
$$ \left(\frac{\mu_{tr}}{\rho}\right)_{\text{mús}} = 0.10(0.0241) + 0.13(0.0206) + 0.73(0.0395) \approx \mathbf{0.0339\text{ cm}^2/\text{g}} $$

##### **3. Coeficientes del Hueso ($w_H=0.06, w_C=0.29, w_O=0.41, w_{Ca}=0.15$):**
$$ \left(\frac{\mu}{\rho}\right)_{\text{hue}} = 0.06(0.3372) + 0.29(0.1772) + 0.41(0.1961) + 0.15(1.0238) \approx \mathbf{0.3056\text{ cm}^2/\text{g}} $$
$$ \left(\frac{\mu_{tr}}{\rho}\right)_{\text{hue}} = 0.06(0.0241) + 0.29(0.0206) + 0.41(0.0395) + 0.15(0.7769) \approx \mathbf{0.1401\text{ cm}^2/\text{g}} $$

---

#### **Análisis del Contraste Obtenido**

1. **Cociente de coeficientes másicos:**
   $$ \frac{(\mu/\rho)_{\text{hueso}}}{(\mu/\rho)_{\text{músculo}}} = \frac{0.3056}{0.1999} \approx \mathbf{1.53} $$

2. **Cociente de coeficientes lineales (atenuación física real):**
   Calculamos el coeficiente lineal $\mu = (\mu/\rho) \cdot \rho$ para ambos tejidos:
   *   $\mu_{\text{músculo}} = 0.1999\text{ cm}^2/\text{g} \times 1.04\text{ g/cm}^3 \approx 0.2079\text{ cm}^{-1}$
   *   $\mu_{\text{hueso}} = 0.3056\text{ cm}^2/\text{g} \times 1.65\text{ g/cm}^3 \approx 0.5042\text{ cm}^{-1}$

   $$ \frac{\mu_{\text{hueso}}}{\mu_{\text{músculo}}} = \frac{0.5042}{0.2079} \approx \mathbf{2.43} $$

**Conclusión Física:** Aunque la diferencia por gramo de tejido ($\mu/\rho$) es moderada (~$1.53$), al tener el hueso una densidad física notablemente mayor, **el haz de rayos X se atenúa casi 2.5 veces más rápido al atravesar el hueso que al atravesar el músculo**. Esto explica por qué el hueso absorbe gran parte de la radiación proyectando una silueta clara en la placa radiográfica (el hueso se ve "blanco" por la falta de exposición de fotones en la película), proporcionando un **excelente contraste radiológico** a energías de diagnóstico del orden de los $50\text{ keV}$.


```python
# Configuración estética para Colab
sns.set_theme(style='whitegrid', palette='colorblind', font='DejaVu Sans')

# 1. CONSTANTES Y DATOS DE ENTRADA
E0_keV = 50.0
NA = 6.0221415e23  # Número de Avogadro

# Datos Compton (por electrón)
s_sc = 0.52
s_tr = 0.04
s_Compton_elec = s_sc + s_tr  # 0.56 barns/electrón

# Densidades (g/cm^3)
rho_musc = 1.04
rho_hueso = 1.65

# Datos atómicos de los elementos: [Z, A, sigma_coh (barns), sigma_ph (barns)]
elementos_data = {
    'H':  [1, 1.0,  0.0, 0.0],
    'C':  [6, 12.0, 0.0, 0.17],
    'O':  [8, 16.0, 0.0, 0.73],
    'Ca': [20, 40.0, 6.0, 50.8]
}

# Composición en peso (fracciones)
w_musculo = {'H': 0.10, 'C': 0.13, 'O': 0.73, 'Ca': 0.0}
w_hueso   = {'H': 0.06, 'C': 0.29, 'O': 0.41, 'Ca': 0.15}

# 2. CÁLCULO DE COEFICIENTES ELEMENTALES
mu_rho_elem = {}
mu_tr_rho_elem = {}

for elem, (Z, A, s_coh, s_ph) in elementos_data.items():
    s_inc = Z * s_Compton_elec
    s_total = s_coh + s_ph + s_inc
    s_tr_total = s_ph + (Z * s_tr)

    # Conversión a cm^2/g: sigma_barns * (NA / A) * 10^-24
    mu_rho_elem[elem] = s_total * (NA / A) * 1e-24
    mu_tr_rho_elem[elem] = s_tr_total * (NA / A) * 1e-24

# 3. CÁLCULO DE COEFICIENTES MACROSCÓPICOS
mu_rho_musc = sum(w_musculo[el] * mu_rho_elem[el] for el in w_musculo)
mu_rho_hue  = sum(w_hueso[el] * mu_rho_elem[el] for el in w_hueso)

mu_tr_rho_musc = sum(w_musculo[el] * mu_tr_rho_elem[el] for el in w_musculo)
mu_tr_rho_hue  = sum(w_hueso[el] * mu_tr_rho_elem[el] for el in w_hueso)

# Coeficientes lineales (cm^-1)
mu_musc = mu_rho_musc * rho_musc
mu_hue  = mu_rho_hue * rho_hueso

# 4. IMPRESIÓN DE RESULTADOS
print("=====================================================================")
print("                        RESOLUCIÓN NUMÉRICA                          ")
print("=====================================================================")
print(f"a) Calcio (Z=20) a 50 keV (Idéntico a Problema 6):")
s_total_Ca = elementos_data['Ca'][1] + elementos_data['Ca'][2] + (20 * s_Compton_elec)
E_tr_Ca = (E0_keV / s_total_Ca) * (elementos_data['Ca'][2] + 20 * s_tr)
print(f"   -> Energía media transferida (E_tr): {E_tr_Ca:.3f} keV")
print(f"   -> Energía media absorbida (E_ab):    {E_tr_Ca:.3f} keV")
print("-" * 69)
print(f"b) Coeficientes de interacción calculados:")
print(f"   -> MÚSCULO:  μ/ρ = {mu_rho_musc:.4f} cm²/g  |  μ = {mu_musc:.4f} cm⁻¹")
print(f"                μ_tr/ρ = {mu_tr_rho_musc:.4f} cm²/g")
print(f"   -> HUESO:    μ/ρ = {mu_rho_hue:.4f} cm²/g  |  μ = {mu_hue:.4f} cm⁻¹")
print(f"                μ_tr/ρ = {mu_tr_rho_hue:.4f} cm²/g")
print("-" * 69)
print("Cocientes de contraste radiológico:")
print(f"   -> Relación de coeficientes másicos:  {mu_rho_hue/mu_rho_musc:.3f}")
print(f"   -> Relación de coeficientes lineales: {mu_hue/mu_musc:.3f} (Contraste real)")
print("=====================================================================\n")

# 5. GENERACIÓN DEL GRÁFICO COMPARATIVO
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle('Análisis del Contraste Radiológico: Músculo vs Hueso a 50 keV',
             fontsize=15, fontweight='bold', y=1.05)

# Gráfico 1: Coeficientes de masa (másicos)
df_masa = pd.DataFrame({
    'Tejido': ['Músculo', 'Hueso', 'Músculo', 'Hueso'],
    'Coeficiente': ['Total (μ/ρ)', 'Total (μ/ρ)', 'Transferencia (μ_tr/ρ)', 'Transferencia (μ_tr/ρ)'],
    'Valor (cm²/g)': [mu_rho_musc, mu_rho_hue, mu_tr_rho_musc, mu_tr_rho_hue]
})
sns.barplot(data=df_masa, x='Coeficiente', y='Valor (cm²/g)', hue='Tejido', ax=ax1)
ax1.set_title('Coeficientes Másicos (Independientes de la densidad)', fontsize=12)
ax1.set_ylabel('Coeficiente másico ($cm^2/g$)', fontsize=11)
ax1.set_xlabel('')
for c in ax1.containers:
    ax1.bar_label(c, fmt='%.3f', fontweight='bold', padding=3)

# Gráfico 2: Coeficiente lineal (atenuación espacial)
sns.barplot(x=['Músculo\n(ρ = 1.04)', 'Hueso\n(ρ = 1.65)'], y=[mu_musc, mu_hue],
            palette=['#4c72b0', '#c44e52'], ax=ax2)
ax2.set_title('Coeficiente Lineal (Atenuación real por cm recorridos)', fontsize=12)
ax2.set_ylabel('Coeficiente lineal μ ($cm^{-1}$)', fontsize=11)
ax2.text(0.5, 0.4, f'El hueso atenúa\n{mu_hue/mu_musc:.2f} veces más\npor cm que el músculo',
         ha='center', va='center', color='darkred', fontsize=11, fontweight='bold',
         bbox=dict(facecolor='white', alpha=0.9, edgecolor='gray', boxstyle="round,pad=0.5"))
ax2.set_ylim(0, 0.6)
for c in ax2.containers:
    ax2.bar_label(c, fmt='%.3f', fontweight='bold', padding=3)

sns.despine(fig=fig)
plt.tight_layout()
plt.show()
```

### **Problema 11:**

Empleando los datos de interacción de rayos $\gamma$ de $30\text{ MeV}$ en plomo, calcule:

* **a)** El **camino libre medio** expresado en unidades de espesor másico ($\lambda_{\rho}$) y en longitud física ($\lambda$).
* **b)** La **probabilidad de que un fotón interaccione** al atravesar perpendicularmente un espesor de $1\text{ cm}$ de plomo.
* **c)** El **valor medio de la energía transferida** por colisión ($\bar{E}_{tr}$).
* **d)** Por separado, el **valor medio de la energía de los electrones** ($\bar{E}_{e^-}$) **y de los positrones** ($\bar{E}_{e^+}$) emitidos, sabiendo que en el proceso de creación de pares el electrón y el positrón se llevan, en media, la misma energía.
* **e)** El **número medio de electrones** ($\bar{n}_{e^-}$) **y de positrones** ($\bar{n}_{e^+}$) por separado que un fotón genera o arranca al atravesar $1\text{ cm}$ de plomo.

**DATOS:**
* Los únicos procesos que tienen lugar son el scattering incoherente y la creación de pares, cuyas secciones eficaces son $\sigma_{inc} = 1.8\text{ barn}$ y $\sigma_{par} = 22.2\text{ barn}$ respectivamente.
* Las secciones eficaces de transferencia y scattering para el efecto Compton son $\sigma_{tr} = 0.0164\text{ barn}$ y $\sigma_{sc} = 0.0056\text{ barn}$ por electrón.
* Densidad del plomo: $\rho = 11.4\text{ g}\cdot\text{cm}^{-3}$.
* Peso atómico medio del plomo: $P_a = 207.2\text{ g/mol}$.
* Número de Avogadro: $N_A = 6.02 \times 10^{23}\text{ átomos/mol}$.
* $1\text{ barn} = 10^{-24}\text{ cm}^2$.


---

#### **a) Camino libre medio ($\lambda_{\rho}$ y $\lambda$)**

1. **Sección eficaz total de interacción ($\sigma$):**
   A la energía de $30\text{ MeV}$, los procesos de interacción considerados son el scattering incoherente y la creación de pares:
   $$ \sigma = \sigma_{inc} + \sigma_{par} = 1.8\text{ b} + 22.2\text{ b} = 24.0\text{ barns} = 24.0 \times 10^{-24}\text{ cm}^2 $$

2. **Coeficiente de atenuación másico ($\mu/\rho$):**
   $$ \frac{\mu}{\rho} = \sigma \cdot \frac{N_A}{P_a} $$
   $$ \frac{\mu}{\rho} = \left(24.0 \times 10^{-24}\text{ cm}^2\right) \times \frac{6.02 \times 10^{23}\text{ mol}^{-1}}{207.2\text{ g/mol}} \approx 0.06973\text{ cm}^2/\text{g} $$

3. **Camino libre medio másico ($\lambda_{\rho}$):**
   Es el recíproco del coeficiente de atenuación másico:
   $$ \lambda_{\rho} = \left(\frac{\mu}{\rho}\right)^{-1} = \frac{1}{0.06973\text{ cm}^2/\text{g}} \approx 14.34\text{ g/cm}^2 $$

4. **Camino libre medio físico ($\lambda$):**
   Dividiendo el espesor másico entre la densidad física del plomo ($\rho = 11.4\text{ g/cm}^3$):
   $$ \lambda = \frac{\lambda_{\rho}}{\rho} = \frac{14.34\text{ g/cm}^2}{11.4\text{ g/cm}^3} \approx 1.26\text{ cm} $$

---

#### **b) Probabilidad de interacción en $1\text{ cm}$ de plomo**

La probabilidad de que un fotón interaccione al atravesar una distancia lineal $x = 1\text{ cm}$ viene dada por la ley exponencial:
$$ P = 1 - e^{-x/\lambda} $$
$$ P = 1 - e^{-1/1.258} \approx 1 - 0.451 = 0.549 \quad (54.9\%) $$

---

#### **c) Valor medio de la energía transferida por colisión ($\bar{E}_{tr}$)**

La energía media transferida al medio por colisión se obtiene como el promedio ponderado de la energía transferida en cada proceso:
$$ \bar{E}_{tr} = \bar{E}_{tr}^{inc} \frac{\sigma_{inc}}{\sigma} + \bar{E}_{tr}^{par} \frac{\sigma_{par}}{\sigma} $$

*   **Scattering Incoherente (Compton):**
    La fracción de energía transferida al electrón de retroceso en el efecto Compton viene dada por:
  $$ \bar{E}_{tr}^{inc} = E_0 \cdot \frac{\sigma_{tr}}{\sigma_{sc} + \sigma_{tr}} = 30\text{ MeV} \times \frac{0.0164}{0.0056 + 0.0164} = 30\text{ MeV} \times 0.7455 \approx 22.36\text{ MeV} $$

*   **Creación de Pares:**
    El fotón desaparece transfiriendo toda su energía excepto la requerida para materializar el par electrón-positrón
  ($2 m_e c^2 \approx 1.022\text{ MeV}$):
  $$ \bar{E}_{tr}^{par} = E_0 - 2 m_e c^2 = 30\text{ MeV} - 1.022\text{ MeV} = 28.98\text{ MeV} $$

Sustituyendo los valores:
$$ \bar{E}_{tr} = 22.36\text{ MeV} \times \left(\frac{1.8}{24.0}\right) + 28.98\text{ MeV} \times \left(\frac{22.2}{24.0}\right) \approx 1.68\text{ MeV} + 26.81\text{ MeV} = 28.5\text{ MeV} $$

---

#### **d) Energía media de electrones ($\bar{E}_{e^-}$) y positrones ($\bar{E}_{e^+}$) emitidos**

Es preciso distinguir las condiciones de emisión para cada tipo de partícula cargada:

*   **Positrones ($\bar{E}_{e^+}$):**
    Solo se producen en la creación de pares. Su energía media cuando son emitidos es la mitad de la energía cinética excedente de ese proceso:
  $$ \bar{E}_{e^+} = \frac{1}{2} \left(E_0 - 2 m_e c^2\right) = \frac{28.98\text{ MeV}}{2} = 14.5\text{ MeV} $$

*   **Electrones ($\bar{E}_{e^-}$):**
    Se emiten tanto en interacciones Compton como en creación de pares. Su energía media es la media ponderada según la probabilidad relativa de cada proceso:
  $$ \bar{E}_{e^-} = \frac{\sigma_{inc}}{\sigma} \bar{E}_{tr}^{inc} + \frac{\sigma_{par}}{\sigma} \left[ \frac{1}{2}\left(E_0 - 2 m_e c^2\right) \right] $$
  $$ \bar{E}_{e^-} = \left(\frac{1.8}{24.0}\right) \times 22.36\text{ MeV} + \left(\frac{22.2}{24.0}\right) \times 14.49\text{ MeV} \approx 1.68 + 13.40 = 15.1\text{ MeV} $$

---

#### **e) Número medio de electrones ($\bar{n}_{e^-}$) y positrones ($\bar{n}_{e^+}$) generados en $1\text{ cm}$**

1. **Número medio de positrones ($\bar{n}_{e^+}$):**
   Se genera exactamente un positrón si ocurre el proceso de creación de pares. Multiplicamos la probabilidad de interacción en $1\text{ cm}$ por la fracción correspondiente a este proceso:
   $$ \bar{n}_{e^+} = P \times \frac{\sigma_{par}}{\sigma} = 0.549 \times \frac{22.2}{24.0} \approx 0.508 \text{ positrones} $$

2. **Número medio de electrones ($\bar{n}_{e^-}$):**
   Tanto el efecto Compton como la creación de pares producen exactamente un electrón por interacción. Por lo tanto, cualquier evento de interacción genera un electrón, de modo que el número medio de electrones coincide con la probabilidad de interacción total:
   $$ \bar{n}_{e^-} = P \times \left(1 \cdot \frac{\sigma_{inc}}{\sigma} + 1 \cdot \frac{\sigma_{par}}{\sigma}\right) = P \approx 0.549 \text{ electrones} $$

```python
# 1. DATOS DE ENTRADA
E0_MeV = 30.0
me_c2 = 0.511
rho_Pb = 11.4      # g/cm^3
Pa_Pb = 207.2      # g/mol
NA = 6.02e23       # Constante de Avogadro

# Secciones eficaces en barns
s_inc = 1.8
s_par = 22.2

# Compton por electrón libre
s_tr = 0.0164
s_sc = 0.0056

# =====================================================================
# CÁLCULOS FÍSICOS
# =====================================================================

# a) Sección eficaz total y camino libre medio
s_total = s_inc + s_par  # barns

# mu/rho en cm^2/g
mu_rho = s_total * 1e-24 * (NA / Pa_Pb)
lambda_rho = 1.0 / mu_rho  # g/cm^2
lambda_fisico = lambda_rho / rho_Pb  # cm

# b) Probabilidad de interacción en x = 1 cm
x_cm = 1.0
P_interaccion = 1.0 - np.exp(-x_cm / lambda_fisico)

# c) Energía media transferida por colisión (E_tr)
E_tr_Compton = E0_MeV * (s_tr / (s_tr + s_sc))
E_tr_pares = E0_MeV - 2.0 * me_c2
E_tr_colision = (s_inc / s_total) * E_tr_Compton + (s_par / s_total) * E_tr_pares

# d) Energía media de electrones y positrones emitidos
E_electron_Compton = E_tr_Compton
E_electron_pares = E_tr_pares / 2.0

E_positron_medio = E_electron_pares
E_electron_medio = (s_inc / s_total) * E_electron_Compton + (s_par / s_total) * E_electron_pares

# e) Número medio de electrones y positrones en 1 cm de espesor
n_positrones = P_interaccion * (s_par / s_total)
n_electrones = P_interaccion * 1.0

# =====================================================================
# IMPRESIÓN DE RESULTADOS
# =====================================================================
print("=====================================================================")
print("                        RESULTADOS NUMÉRICOS                         ")
print("=====================================================================")
print(f"Sección eficaz total de interacción (σ): {s_total:.2f} barns")
print("-" * 69)
print(f"a) Camino libre medio másico (λ_ρ):      {lambda_rho:.4f} g/cm²")
print(f"   Camino libre medio físico (λ):        {lambda_fisico:.4f} cm")
print("-" * 69)
print(f"b) Probabilidad de interacción en 1 cm:  {P_interaccion*100:.3f}%")
print("-" * 69)
print(f"c) Energía media transferida (E_tr):     {E_tr_colision:.4f} MeV")
print(f"   └─ Componente Compton: {E_tr_Compton:.4f} MeV | Componente Pares: {E_tr_pares:.4f} MeV")
print("-" * 69)
print(f"d) Energía media de partículas emitidas:")
print(f"   -> Positrones (E_e+):                 {E_positron_medio:.4f} MeV")
print(f"   -> Electrones (E_e-):                 {E_electron_medio:.4f} MeV")
print("-" * 69)
print(f"e) Partículas generadas al atravesar 1 cm de Plomo:")
print(f"   -> Positrones medios (n_e+):          {n_positrones:.4f}")
print(f"   -> Electrones medios (n_e-):          {n_electrones:.4f}")
print("=====================================================================")
```

### **Problema 12**

Un haz de fotones de $15\text{ MeV}$ incide sobre un cierto material monoatómico. Los únicos procesos que tienen lugar son el scattering incoherente y la creación de pares, que para esta energía tienen la misma sección eficaz. Resolver las siguientes cuestiones:

* **a)** Sabiendo que el haz necesita atravesar un espesor de $11.2\text{ g}\cdot\text{cm}^{-2}$ de ese material para atenuarse un $10\%$ ($1.0 \to 0.90$), calcule la **sección eficaz total** de interacción.
* **b)** Halle la **energía media transferida** por colisión ($\bar{E}_{tr}$) y la **energía que un fotón transfiere en media** al atravesar ese espesor.
* **c)** Calcule por separado el **valor medio de la energía de los electrones** ($\bar{E}_{e^-}$) **y de los positrones** ($\bar{E}_{e^+}$) que se producen.
* **d)** Describa de manera analítica y cuantitativa la forma de los **espectros de energía de los positrones y de los electrones**, indicando sus límites y características clave.
* **e)** Para un haz que se mueve en un medio infinito, halle la **fracción de la energía total** que se transforma en energía cinética de electrones ($f_{e^-}$) y en energía cinética de positrones ($f_{e^+}$). ¿En qué se invierte el resto de la energía?

**Datos:**
* Densidad atómica del material: $\rho_A = 9.48 \times 10^{21}\text{ átomos/gramo}$.
* A $15\text{ MeV}$, las secciones eficaces Compton (por electrón) de scattering y transferencia valen: $\sigma_{sc} = 0.0109\text{ barn}$ y $\sigma_{tr} = 0.0268\text{ barn}$ respectivamente.
* Energía en reposo del electrón/positrón: $m_e c^2 = 0.511\text{ MeV}$.
* $1\text{ barn} = 10^{-24}\text{ cm}^2$.

---

#### **a) Sección eficaz total de interacción ($\sigma$)**

1. **Coeficiente de atenuación másico ($\mu/\rho$):**
   La atenuación sigue la ley exponencial $I = I_0 e^{-\left(\frac{\mu}{\rho}\right) t}$. Sabiendo que el haz se atenúa un $10\%$, la intensidad transmitida es $I/I_0 = 0.90$ para un espesor de masa $t = 11.2\text{ g}\cdot\text{cm}^{-2}$:
   $$ \frac{\mu}{\rho} = -\frac{1}{t} \ln\left(\frac{I}{I_0}\right) = -\frac{\ln(0.90)}{11.2\text{ g/cm}^2} \approx 9.407 \times 10^{-3}\text{ cm}^2/\text{g} $$

2. **Sección eficaz total por átomo ($\sigma$):**
   El coeficiente de atenuación másico se relaciona con la sección eficaz atómica total mediante $\mu/\rho = \rho_A \cdot \sigma$, donde $\rho_A$ es la densidad atómica en átomos/gramo:
   $$ \sigma = \frac{\mu/\rho}{\rho_A} = \frac{9.407 \times 10^{-3}\text{ cm}^2/\text{g}}{9.48 \times 10^{21}\text{ átomos/g}} \approx 9.923 \times 10^{-25}\text{ cm}^2 = \mathbf{0.992\text{ barns}} $$

Como el enunciado indica que ambos procesos tienen la misma sección eficaz a esta energía:
$$ \sigma_{inc} = \sigma_{par} = \frac{1}{2} \sigma \approx \mathbf{0.496\text{ barns}} $$

---

#### **b) Energía media transferida por colisión ($\bar{E}_{tr}$) y transferida en la placa**

1. **Energía media transferida Compton ($\bar{E}_{tr}^{inc}$):**
   La sección eficaz Compton total por electrón es la suma de sus componentes: $\sigma_{KN} = \sigma_{sc} + \sigma_{tr} = 0.0109 + 0.0268 = 0.0377\text{ barns}$.
   La energía transferida de media al electrón Compton es:
   $$ \bar{E}_{tr}^{inc} = E_0 \frac{\sigma_{tr}}{\sigma_{sc} + \sigma_{tr}} = 15\text{ MeV} \times \frac{0.0268}{0.0377} \approx 10.66\text{ MeV} $$

2. **Energía media transferida en creación de pares ($\bar{E}_{tr}^{par}$):**
   Toda la energía del fotón, excepto la empleada en la masa en reposo de las dos partículas ($2 m_e c^2 \approx 1.022\text{ MeV}$), se transfiere como energía cinética:
   $$ \bar{E}_{tr}^{par} = E_0 - 2 m_e c^2 = 15 - 1.022 = 13.98\text{ MeV} $$

3. **Energía media transferida por colisión total ($\bar{E}_{tr}$):**
   Ponderamos las transferencias según la probabilidad relativa de cada proceso ($\sigma_i/\sigma = 0.5$):
   $$ \bar{E}_{tr} = 0.5 \cdot \bar{E}_{tr}^{inc} + 0.5 \cdot \bar{E}_{tr}^{par} $$
   $$ \bar{E}_{tr} = 0.5 \times (10.66\text{ MeV}) + 0.5 \times (13.98\text{ MeV}) \approx \mathbf{12.32\text{ MeV}} $$

4. **Energía transferida en media por fotón incidente que atraviesa la placa ($\langle E \rangle_{tr}$):**
   Dado que solo interacciona el $10\%$ de los fotones ($P_{int} = 0.10$):
   $$ \langle E \rangle_{tr} = P_{int} \cdot \bar{E}_{tr} = 0.10 \times 12.32\text{ MeV} \approx \mathbf{1.23\text{ MeV}} $$

---

#### **c) Energía media de electrones ($\bar{E}_{e^-}$) y positrones ($\bar{E}_{e^+}$) producidos**

*   **Positrones ($\bar{E}_{e^+}$):**
    Solo se producen en la creación de pares y, según las indicaciones del enunciado, el electrón y el positrón se reparten la energía cinética excedente de forma equitativa en promedio:
    $$ \bar{E}_{e^+} = \frac{1}{2} \left( E_0 - 2 m_e c^2 \right) = \frac{13.98\text{ MeV}}{2} \approx \mathbf{6.99\text{ MeV}} $$

*   **Electrones ($\bar{E}_{e^-}$):**
    Se producen tanto en el efecto Compton como en la creación de pares. Dado que ambos procesos ocurren con igual sección eficaz (
  $\sigma_{inc} = \sigma_{par}$), las probabilidades de generar un electrón en cada canal son del $50\%$:
  $$ \bar{E}_{e^-} = 0.5 \cdot \bar{E}_{tr}^{inc} + 0.5 \cdot \left[ \frac{1}{2} \left( E_0 - 2 m_e c^2 \right) \right] $$
  $$ \bar{E}_{e^-} = 0.5 \times (10.66\text{ MeV}) + 0.5 \times (6.99\text{ MeV}) \approx \mathbf{8.83\text{ MeV}} $$

---

#### **d) Espectroscopia cuantitativa de las partículas emitidas**

*   **Espectro de los Positrones ($e^+$):**
    Los positrones provienen únicamente del canal de creación de pares. Es una **distribución continua y simétrica**, centrada en su valor medio de **$6.99\text{ MeV}$**. El espectro arranca en $0\text{ MeV}$ y se extiende hasta una energía cinética máxima de:
    $$ E_{max}^{e^+} = E_0 - 2 m_e c^2 \approx \mathbf{13.98\text{ MeV}} $$

*   **Espectro de los Electrones ($e^-$):**
    Es el resultado de la **superposición** de dos componentes físicas diferenciadas:
    1. Una **contribución de pares**: Curva continua y simétrica idéntica a la de los positrones, centrada en $6.99\text{ MeV}$ con límite en $13.98\text{ MeV}$.
    2. Una **contribución Compton**: Espectro asimétrico característico que empieza en $0\text{ MeV}$ y crece progresivamente hasta alcanzar un máximo abrupto en el denominado **borde Compton** ($\gamma = E_0/m_e c^2 = 15/0.511 \approx 29.35$):
   $$ E_{max}^{inc} = E_0 \frac{2\gamma}{1 + 2\gamma} = 15\text{ MeV} \times \frac{58.7}{59.7} \approx \mathbf{14.75\text{ MeV}} $$
    La suma de ambas curvas resulta en una distribución combinada con un valor medio de **$8.83\text{ MeV}$** y un límite físico superior de **$14.75\text{ MeV}$**.

---

#### **e) Balance de energía en un medio infinito**

En un medio infinito todo fotón interacciona ($P_{int} = 1.0$), por lo que podemos rastrear el destino del 100% de la energía del haz incidente ($15\text{ MeV}$):

1. **Fracción en energía cinética de positrones ($f_{e^+}$):**
   $$ f_{e^+} = \frac{0.5 \cdot \bar{E}_{e^+}}{E_0} = \frac{0.5 \times 6.99\text{ MeV}}{15\text{ MeV}} \approx \mathbf{23.3\%} $$

2. **Fracción en energía cinética de electrones ($f_{e^-}$):**
   $$ f_{e^-} = \frac{0.5 \cdot \bar{E}_{tr}^{inc} + 0.5 \cdot \bar{E}_{e^-, par}}{E_0} = \frac{8.83\text{ MeV}}{15\text{ MeV}} \approx \mathbf{58.8\%} $$

3. **Inversión de la energía restante:**
   El resto de la energía total ($100\% - 23.3\% - 58.8\% = \mathbf{17.9\%}$) se distribuye de la siguiente manera:
   *   **Fotones dispersados por efecto Compton ($f_{\gamma}^{sc}$):** Transportan la energía remanente de los choques incoherentes.
  $$ f_{\gamma}^{sc} = \frac{0.5 \cdot \bar{E}_{sc}^{inc}}{E_0} = 0.5 \times \frac{\sigma_{sc}}{\sigma_{KN}} = 0.5 \times \frac{0.0109}{0.0377} \approx \mathbf{14.5\%} $$
   *   **Materialización de masa en reposo del par ($f_{masa}$):** Energía invertida en crear la masa de los electrones y positrones.
  $$ f_{masa} = \frac{0.5 \cdot (2 m_e c^2)}{E_0} = 0.5 \times \frac{1.022\text{ MeV}}{15\text{ MeV}} \approx \mathbf{3.4\%} $$

```python

# 1. DATOS DE ENTRADA
E0_MeV = 15.0
me_c2 = 0.511
t_espesor = 11.2       # g/cm^2
I_over_I0 = 0.90       # 10% de atenuación
rho_A = 9.48e21        # átomos/gramo

# Compton (por electrón libre)
s_sc = 0.0109          # barns
s_tr = 0.0268          # barns

# =====================================================================
# DESARROLLO DE CÁLCULOS
# =====================================================================

# --- Apartado a) Sección eficaz total ---
mu_rho = -np.log(I_over_I0) / t_espesor  # cm^2/g
sigma_total_cm2 = mu_rho / rho_A
sigma_total_barns = sigma_total_cm2 / 1e-24

# --- Apartado b) Energía media transferida ---
s_KN = s_sc + s_tr
E_tr_inc = E0_MeV * (s_tr / s_KN)
E_tr_par = E0_MeV - 2.0 * me_c2

# Ponderación 50% / 50%
E_tr_colision = 0.5 * E_tr_inc + 0.5 * E_tr_par
E_tr_medio_haz = E_tr_colision * (1.0 - I_over_I0)

# --- Apartado c) Energías por partícula ---
E_pos_medio = 0.5 * E_tr_par
E_ele_medio = 0.5 * E_tr_inc + 0.5 * E_pos_medio

# --- Apartado d) Espectroscopia (Borde Compton) ---
gamma = E0_MeV / me_c2
E_borde_Compton = E0_MeV * ((2.0 * gamma) / (1.0 + 2.0 * gamma))

# --- Apartado e) Balance energético en medio infinito ---
f_pos = 0.5 * E_pos_medio / E0_MeV
f_ele = E_ele_medio / E0_MeV
f_sc_gamma = 0.5 * (s_sc / s_KN)
f_mass = 0.5 * (2.0 * me_c2) / E0_MeV

# =====================================================================
# IMPRESIÓN DE RESULTADOS
# =====================================================================
print("=====================================================================")
print("                RESULTADOS NUMÉRICOS (PROBLEMA 12)                   ")
print("=====================================================================")
print(f"a) Coeficiente másico de atenuación: {mu_rho:.5e} cm²/g")
print(f"   Sección eficaz total (σ):        {sigma_total_barns:.4f} barns")
print(f"   └─ Componente Compton (σ_inc):   {sigma_total_barns/2.0:.4f} barns")
print(f"   └─ Componente Pares (σ_par):     {sigma_total_barns/2.0:.4f} barns")
print("-" * 69)
print(f"b) Energía media transferida/colisión (E_tr): {E_tr_colision:.4f} MeV")
print(f"   Energía media transferida en la placa:     {E_tr_medio_haz:.4f} MeV")
print("-" * 69)
print(f"c) Energía media de partículas producidas:")
print(f"   -> Positrones (E_e+):                      {E_pos_medio:.4f} MeV")
print(f"   -> Electrones (E_e-):                      {E_ele_medio:.4f} MeV")
print("-" * 69)
print(f"d) Parámetros de espectroscopia:")
print(f"   -> Energía máxima de positrones (E_max_e+): {E_tr_par:.4f} MeV")
print(f"   -> Borde Compton de electrones (E_max_inc): {E_borde_Compton:.4f} MeV")
print("-" * 69)
print(f"e) Fracciones de balance energético (Medio Infinito):")
print(f"   -> En energía cinética de electrones (f_e-): {f_ele*100:6.2f}%")
print(f"   -> En energía cinética de positrones (f_e+): {f_pos*100:6.2f}%")
print(f"   -> En energía de fotones dispersados (f_sc): {f_sc_gamma*100:6.2f}%")
print(f"   -> En materialización de masa en reposo:     {f_mass*100:6.2f}%")
print(f"   Suma de fracciones de balance:               {(f_pos+f_ele+f_sc_gamma+f_mass)*100:.2f}%")
print("=====================================================================")
```

#### **Justificación de los espectros de electrones y positrones (Apartado d)**

El siguiente gráfico  representa las **distribuciones continuas de probabilidad teóricas puras** (sin ensanchamiento instrumental por resolución de detector) para las partículas cargadas generadas por el haz de fotones de $15\text{ MeV}$:

*   **Espectro del Positrón ($e^+$, Panel Inferior):** Gobernado exclusivamente por el canal de **creación de pares (Bethe-Heitler)**. Para energías relativistas amplias (como $15\text{ MeV}$), la sección eficaz diferencial para la creación de pares como función de la fracción de energía que se lleva el positrón ($u = T_+ / E_{\text{max,par}}$) se modela de forma clásica como: $$ \frac{d\sigma_{par}}{dT_+} \propto 1 - \frac{4}{3}u(1-u) $$ Esta ecuación describe una curva simétrica en forma de "U" (o copa). Presenta un mínimo local en el centro ($u = 0.5$, correspondiente a un reparto equitativo de $6.99\text{ MeV}$) y asciende de forma asintótica hacia los extremos ($u \to 0$ y $u \to 1$). Su silueta simétrica en forma de "U" demuestra cuánticamente que es más probable un reparto de energía asimétrico entre el par (donde un miembro se lleva casi toda la energía y el otro nace casi en reposo) que un reparto equitativo del $50/50$. La distribución se interrumpe de forma vertical en su límite cinemático de **$13.98\text{ MeV}$** ($E_0 - 2m_e c^2$).

*   **Espectro del Electrón ($e^-$, Panel Superior):** Resulta de la superposición exacta (ponderada al $50\%$ por igualdad de sección eficaz) del canal de creación de pares y del scattering Compton (**Klein-Nishina**). La ausencia de resolución instrumental permite identificar con total nitidez las dos caídas verticales teóricas (*cliffs*) del sistema:
    1.  A **$13.98\text{ MeV}$**, donde se extingue la componente de creación de pares.
    2.  A **$14.75\text{ MeV}$**, correspondiente al límite físico de transferencia por colisión frontal o **Borde Compton**.


```python
# -----------------------------------------------------------------
# 1. PARÁMETROS FÍSICOS (Fotones incidentes de 15 MeV)
# -----------------------------------------------------------------
E0 = 15.0      # Energía del fotón incidente (MeV)
me_c2 = 0.511  # Energía en reposo del electrón (MeV)
alpha = E0 / me_c2

# Límites cinéticos teóricos
E_max_par = E0 - 2.0 * me_c2  # Creación de Pares: 13.978 MeV
E_max_compton = E0 * (2.0 * alpha) / (1.0 + 2.0 * alpha)  # Borde Compton: 14.755 MeV

# Eje de energías de simulación (de 0 a 15.5 MeV)
N_puntos = 2000
x = np.linspace(0.001, 15.5, N_puntos)
dx = x[1] - x[0]  # Paso real decimal de la rejilla de integración

# -----------------------------------------------------------------
# 2. MODELADO DE LAS DISTRIBUCIONES TEÓRICAS (PDFs NORMALIZADAS)
# -----------------------------------------------------------------

# A. Creación de Pares (Distribución de Bethe-Heitler en forma de "U" pura)
def pair_pdf_bh(E):
    if E <= 0 or E > E_max_par:
        return 0.0
    u = E / E_max_par
    return 1.0 - (4.0 / 3.0) * u * (1.0 - u)

pair_vals = np.array([pair_pdf_bh(e) for e in x])
pair_pdf_norm = pair_vals / (np.sum(pair_vals) * dx)

# B. Scattering Compton (Sección eficaz diferencial dSigma/dT de Klein-Nishina con Jacobiano)
def compton_pdf_correct(T):
    if T <= 0 or T > E_max_compton:
        return 0.0
    Ep = E0 - T  # Energía del fotón dispersado (E')
    cos_theta = 1.0 - (1.0 / alpha) * (T / Ep)
    sin_sq_theta = np.clip(1.0 - cos_theta**2, 0.0, 1.0)

    # Ecuación correcta de dSigma/dT tras simplificar el Jacobiano de cambio de variable
    bracket = Ep / E0 + E0 / Ep - sin_sq_theta
    return bracket

compton_vals = np.array([compton_pdf_correct(t) for t in x])
compton_pdf_norm = compton_vals / (np.sum(compton_vals) * dx)

# Espectro total de electrones (Superposición equitativa al 50% Compton y 50% Pares)
total_electron_pdf = 0.5 * compton_pdf_norm + 0.5 * pair_pdf_norm

# -----------------------------------------------------------------
# 3. RENDERIZADO DE LAS GRÁFICAS APILADAS (STACKED LAYOUT)
# -----------------------------------------------------------------
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 11), sharex=True)

# --- PANEL SUPERIOR: Espectro de Electrones (e-) ---
# Componentes teóricas individuales (líneas discontinuas)
ax1.plot(x, 0.5 * pair_pdf_norm, color='#c44e52', linewidth=2, linestyle='--',
         label=r'Componente de Pares ($e^-_{par}$ - Bethe-Heitler)')
ax1.plot(x, 0.5 * compton_pdf_norm, color='#55a868', linewidth=2, linestyle=':',
         label=r'Componente Compton ($e^-_{inc}$ - Klein-Nishina)')

# Curva combinada total observable
ax1.plot(x, total_electron_pdf, color='#2c3e50', linewidth=3,
         label=r'Espectro Total de Electrones ($e^-$)')
ax1.fill_between(x, total_electron_pdf, color='#2c3e50', alpha=0.08)

# Indicadores de valores medios calculados en el problema
ax1.axvline(x=8.83, color='#4c72b0', linestyle='-.', alpha=0.8, linewidth=1.5)
ax1.text(8.83 + 0.15, np.max(total_electron_pdf) * 0.85, r'$\bar{E}_{e^-} = 8.83$ MeV',
         fontsize=10.5, color='#4c72b0', fontweight='bold')

ax1.axvline(x=10.66, color='#55a868', linestyle=':', alpha=0.8, linewidth=1.5)
ax1.text(10.66 + 0.15, np.max(total_electron_pdf) * 0.65, r'$\bar{E}_{inc} = 10.66$ MeV',
         fontsize=10.5, color='#55a868', fontweight='bold')

ax1.axvline(x=6.99, color='#c44e52', linestyle='--', alpha=0.8, linewidth=1.5)
ax1.text(6.99 - 0.2, np.max(total_electron_pdf) * 0.40, r'$\bar{E}_{e^-, par} = 6.99$ MeV',
         fontsize=10.5, color='#c44e52', fontweight='bold', ha='right')

# Límites cinemáticos abruptos
ax1.axvline(x=E_max_par, color='#c44e52', linestyle='--', alpha=0.5)
ax1.text(E_max_par - 0.15, np.max(total_electron_pdf) * 0.12, f'{E_max_par:.3f} MeV\nLímite de Pares',
         fontsize=9.5, color='#c44e52', ha='right', fontweight='bold')

ax1.axvline(x=E_max_compton, color='#55a868', linestyle=':', alpha=0.5)
ax1.text(E_max_compton + 0.15, np.max(total_electron_pdf) * 0.12, f'{E_max_compton:.3f} MeV\nBorde Compton',
         fontsize=9.5, color='#55a868', ha='left', fontweight='bold')

ax1.set_title('Espectro Teórico de Electrones ($e^-$): Superposición Compton (KN) + Creación de Pares (BH)',
             fontsize=12, fontweight='bold')
ax1.set_ylabel('Densidad de Probabilidad (1/MeV)', fontsize=11)
ax1.set_ylim(0, np.max(total_electron_pdf) * 1.15)
ax1.legend(loc='upper left', fontsize=10)


# --- PANEL INFERIOR: Espectro de Positrones (e+) ---
# Distribución de Bethe-Heitler simétrica pura
ax2.plot(x, pair_pdf_norm, color='#e67e22', linewidth=3,
         label=r'Espectro de Positrones ($e^+$ - Bethe-Heitler)')
ax2.fill_between(x, pair_pdf_norm, color='#e67e22', alpha=0.1)

# Límite cinemático de creación de pares
ax2.axvline(x=E_max_par, color='#e67e22', linestyle='--', alpha=0.5)
ax2.text(E_max_par - 0.15, np.max(pair_pdf_norm) * 0.15, f'Límite de Pares\n{E_max_par:.3f} MeV',
         fontsize=9.5, color='#e67e22', ha='right', fontweight='bold')

ax2.set_title('Espectro Teórico de Positrones ($e^+$): Distribución de Bethe-Heitler Pura (U-Shape)',
             fontsize=12, fontweight='bold')
ax2.set_xlabel('Energía Cinética de la Partícula (MeV)', fontsize=11)
ax2.set_ylabel('Densidad de Probabilidad (1/MeV)', fontsize=11)
ax2.set_xlim(0, 15.5)
ax2.set_ylim(0, np.max(pair_pdf_norm) * 1.15)
ax2.legend(loc='upper center', fontsize=10)

# Limpieza final de contornos
sns.despine(fig=fig)
plt.tight_layout()
plt.show()
```

### **Problema 13:**

Un haz de fotones monoenergético incide sobre una lámina de cierto material de espesor $x = 0.628\text{ g}\cdot\text{cm}^{-2}$. Se sabe que el $4.3\%$ de los fotones sufren scattering coherente, el $5.3\%$ efecto Compton y el $53.6\%$ son absorbidos por efecto fotoeléctrico. El resto de los fotones atraviesan la lámina sin interaccionar. Despreciando efectos de radiación secundaria, determine:

* **a)** El **camino libre medio** ($\lambda$) expresado en unidades de espesor másico ($\text{g}\cdot\text{cm}^{-2}$).
* **b)** Los **coeficientes másicos de atenuación** ($\mu_i/\rho$) para cada uno de los tres procesos considerados.
* **c)** Sabiendo que la máxima energía cinética de los electrones dispersados por efecto Compton es $T_{max} = 421\text{ keV}$, calcule la **energía de los fotones primarios del haz** ($E_0$).

---

#### **a) Camino libre medio másico ($\lambda$)**

1. **Probabilidad total de interacción ($P_{tot}$):**
   La probabilidad de que un fotón sufra algún tipo de interacción al atravesar la lámina es la suma directa de las probabilidades individuales de cada proceso:
   $$ P_{tot} = P_{coh} + P_{inc} + P_{ph} = 0.043 + 0.053 + 0.536 = 0.632 \quad (63.2\%) $$

2. **Cálculo del camino libre medio ($\lambda$):**
   La fracción de fotones que consigue atravesar el espesor másico $x$ sin interaccionar (transmisión) es:
   $$ \frac{I}{I_0} = 1 - P_{tot} = 1 - 0.632 = 0.368 \quad (36.8\%) $$

   Aplicando la ley exponencial de atenuación:
   $$ \frac{I}{I_0} = e^{-x/\lambda} \implies \frac{x}{\lambda} = -\ln(1 - P_{tot}) = -\ln(0.368) \approx 1.000 $$

   Sustituyendo el espesor dado ($x = 0.628\text{ g}\cdot\text{cm}^{-2}$), obtenemos el camino libre medio másico:
   $$ \lambda = \frac{x}{-\ln(1 - P_{tot})} = \frac{0.628\text{ g}\cdot\text{cm}^{-2}}{1.000} = \mathbf{0.628\text{ g}\cdot\text{cm}^{-2}} $$

---

#### **b) Coeficientes másicos de atenuación parciales ($\mu_i/\rho$)**

1. **Coeficiente de atenuación másico total ($\mu/\rho$):**
   Es el recíproco del camino libre medio expresado en espesor másico:
   $$ \frac{\mu}{\rho} = \frac{1}{\lambda} = \frac{1}{0.628\text{ g}\cdot\text{cm}^{-2}} \approx \mathbf{1.592\text{ cm}^2/\text{g}} $$

2. **Coeficientes parciales por proceso ($\mu_i/\rho$):**
   Dado que las secciones eficaces (y por tanto los coeficientes de atenuación) son directamente proporcionales a las fracciones de interacción de cada canal, multiplicamos el coeficiente total por la probabilidad relativa de cada efecto ($P_i / P_{tot}$):
   $$ \frac{\mu_i}{\rho} = \frac{\mu}{\rho} \cdot \frac{P_i}{P_{tot}} $$

   *   **Scattering Coherente:**
       $$ \frac{\mu_{coh}}{\rho} = 1.592\text{ cm}^2/\text{g} \times \frac{0.043}{0.632} \approx \mathbf{0.108\text{ cm}^2/\text{g}} $$
   *   **Scattering Incoherente (Efecto Compton):**
       $$ \frac{\mu_{inc}}{\rho} = 1.592\text{ cm}^2/\text{g} \times \frac{0.053}{0.632} \approx \mathbf{0.133\text{ cm}^2/\text{g}} $$
   *   **Efecto Fotoeléctrico:**
       $$ \frac{\mu_{ph}}{\rho} = 1.592\text{ cm}^2/\text{g} \times \frac{0.536}{0.632} \approx \mathbf{1.350\text{ cm}^2/\text{g}} $$

---

#### **c) Energía de los fotones primarios ($E_0$)**

1. **Ecuación del Borde Compton ($T_{max}$):**
   La máxima energía cinética transferida a un electrón de retroceso Compton ocurre en colisiones frontales ($\theta = 180^\circ$):
   $$ T_{max} = E_0 \left( \frac{2\gamma}{1 + 2\gamma} \right) $$
   Donde $\gamma = E_0 / m_e c^2$ es el parámetro de energía adimensional del fotón incidente (con $m_e c^2 = 511\text{ keV}$).

2. **Definición del parámetro de transferencia máximo ($\gamma_{max}$):**
   Definimos la energía cinética máxima de forma adimensional respecto a la masa del electrón en reposo:
   $$ \gamma_{max} = \frac{T_{max}}{m_e c^2} = \frac{421\text{ keV}}{511\text{ keV}} \approx 0.8239 $$

3. **Planteamiento de la ecuación cuadrática para $\gamma$:**
   Sustituyendo $\gamma_{max}$ en la relación del borde Compton, podemos despejar el valor de $\gamma$:
   $$ \gamma_{max} = \frac{2\gamma^2}{1 + 2\gamma} \implies 2\gamma^2 - 2\gamma_{max}\gamma - \gamma_{max} = 0 $$

4. **Resolución de la ecuación de segundo grado:**
   Tomamos únicamente la raíz físicamente real y positiva:
   $$ \gamma = \frac{2\gamma_{max} + \sqrt{4\gamma_{max}^2 + 8\gamma_{max}}}{4} $$
   $$ \gamma = \frac{2(0.8239) + \sqrt{4(0.8239)^2 + 8(0.8239)}}{4} \approx 1.1746 $$

5. **Determinación de la energía del haz incidente ($E_0$):**
   $$ E_0 = \gamma \cdot m_e c^2 = 1.1746 \times 511\text{ keV} \approx \mathbf{600.2\text{ keV}} \quad (\approx 0.60\text{ MeV}) $$

```python
# 1. DATOS DE ENTRADA DEL PROBLEMA
x_espesor = 0.628     # Espesor másico (g/cm^2)
P_coh = 0.043         # Probabilidad Coherente
P_inc = 0.053         # Probabilidad Compton (Incoherente)
P_ph = 0.536          # Probabilidad Fotoeléctrico
T_max_keV = 421.0     # Energía máxima de retroceso Compton (keV)
me_c2_keV = 511.0     # Energía en reposo del electrón (keV)

# =====================================================================
# CÁLCULOS NUMÉRICOS PRECISOS
# =====================================================================

# a) Camino libre medio (lambda)
P_tot = P_coh + P_inc + P_ph
lambda_val = x_espesor / -np.log(1.0 - P_tot)

# b) Coeficientes parciales de atenuación másica
mu_rho_total = 1.0 / lambda_val
mu_coh_rho = mu_rho_total * (P_coh / P_tot)
mu_inc_rho = mu_rho_total * (P_inc / P_tot)
mu_ph_rho  = mu_rho_total * (P_ph / P_tot)

# c) Energía del fotón primario (E0)
gamma_max = T_max_keV / me_c2_keV

# Coeficientes para la ecuación de segundo grado: 2*gamma^2 - 2*gamma_max*gamma - gamma_max = 0
a = 2.0
b = -2.0 * gamma_max
c = -gamma_max

# Aplicamos la fórmula cuadrática para hallar la raíz positiva
gamma = (-b + np.sqrt(b**2 - 4.0 * a * c)) / (2.0 * a)
E0_keV = gamma * me_c2_keV

# =====================================================================
# IMPRESIÓN DE RESULTADOS FORMATEADOS
# =====================================================================
print("=====================================================================")
print("                RESULTADOS NUMÉRICOS (PROBLEMA 13)                   ")
print("=====================================================================")
print(f"Probabilidad total de interacción (P_tot): {P_tot*100:.1f}%")
print("-" * 69)
print(f"a) Camino libre medio másico (λ):          {lambda_val:.4f} g/cm²")
print("-" * 69)
print(f"b) Coeficientes de atenuación másica parciales (μ_i/ρ):")
print(f"   -> Coeficiente Másico Total:           {mu_rho_total:.4f} cm²/g")
print(f"   -> Scattering Coherente (μ_coh/ρ):     {mu_coh_rho:.4f} cm²/g")
print(f"   -> Scattering Compton (μ_inc/ρ):       {mu_inc_rho:.4f} cm²/g")
print(f"   -> Efecto Fotoeléctrico (μ_ph/ρ):      {mu_ph_rho:.4f} cm²/g")
print("-" * 69)
print(f"c) Energía de los fotones incidentes del haz (E0):")
print(f"   -> Parámetro gamma_max del electrón:   {gamma_max:.4f}")
print(f"   -> Parámetro gamma del fotón primario: {gamma:.4f}")
print(f"   -> Energía de los fotones (E0):        {E0_keV:.2f} keV  (~ {E0_keV/1000.0:.3f} MeV)")
print("=====================================================================")
```

A continuación se presenta una calculadora interactiva de cinemática del proceso Compton

```python
# =====================================================================
# CALCULADORA INTERACTIVA DE CINEMÁTICA COMPTON
# =====================================================================


def simular_compton(E0_keV, theta_deg):
    me_c2_keV = 511.0  # Energía en reposo del electrón (keV)
    alpha = E0_keV / me_c2_keV
    theta_rad = np.radians(theta_deg)

    # 1. Cinemática de la colisión Compton
    # Energía del fotón dispersado (E')
    Ep = E0_keV / (1.0 + alpha * (1.0 - np.cos(theta_rad)))
    # Energía cinética del electrón de retroceso (T)
    T = E0_keV - Ep

    # Ángulo de retroceso del electrón (phi) deducido por conservación de momento
    # p_x = E0 - Ep * cos(theta), p_y = Ep * sin(theta)
    phi_rad = np.arctan2(Ep * np.sin(theta_rad), E0_keV - Ep * np.cos(theta_rad))
    phi_deg = np.degrees(phi_rad)

    # Borde Compton (energía máxima transferible a theta = 180)
    T_max = E0_keV * (2.0 * alpha) / (1.0 + 2.0 * alpha)

    # 2. Renderizado gráfico del doble panel
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
    fig.suptitle(f'Calculadora de Cinemática Compton (E0 = {E0_keV:.1f} keV, θ = {theta_deg}°)',
                 fontsize=14, fontweight='bold', y=0.98)

    # --- PANEL 1: DIAGRAMA VECTORIAL DEL CHOQUE ---
    # Marcador de blanco (electrón libre en reposo)
    ax1.plot(0, 0, 'go', markersize=12, label='Blanco ($e^-$ en reposo)', zorder=5)

    # Fotón incidente: entra por la izquierda a lo largo del eje X
    ax1.quiver(-1.0, 0.0, 1.0, 0.0, angles='xy', scale_units='xy', scale=1,
               color='#4c72b0', width=0.007, label=f'Fotón Incidente ($E_0 = {E0_keV:.1f}$ keV)', zorder=3)

    # Fotón dispersado: sale en el ángulo theta con longitud proporcional a su energía Ep/E0
    len_photon = Ep / E0_keV
    ax1.quiver(0.0, 0.0, len_photon * np.cos(theta_rad), len_photon * np.sin(theta_rad),
               angles='xy', scale_units='xy', scale=1, color='#55a868', width=0.007,
               label=f"Fotón Dispersado ($E' = {Ep:.1f}$ keV)", zorder=4)

    # Electrón de retroceso: sale desviado en el ángulo -phi con longitud proporcional a T/E0
    len_electron = T / E0_keV
    if T > 0:  # Evitar graficar flechas nulas a theta = 0
        ax1.quiver(0.0, 0.0, len_electron * np.cos(-phi_rad), len_electron * np.sin(-phi_rad),
                   angles='xy', scale_units='xy', scale=1, color='#c44e52', width=0.007,
                   label=f'Electrón de Retroceso ($T = {T:.1f}$ keV)', zorder=4)

    # Líneas de referencia angulares (ejes cartesianos discretos)
    ax1.axhline(0, color='gray', linestyle='--', alpha=0.3, zorder=1)
    ax1.axvline(0, color='gray', linestyle='--', alpha=0.3, zorder=1)

    # Ajustes del panel vectorial
    ax1.set_xlim(-1.2, 1.2)
    ax1.set_ylim(-1.2, 1.2)
    ax1.set_aspect('equal')
    ax1.set_title('Diagrama Vectorial del Choque', fontsize=12, fontweight='bold', pad=10)
    ax1.legend(loc='upper left', fontsize=9)
    ax1.set_xlabel('Momento relativo en X ($p_x / p_0$)', fontsize=10)
    ax1.set_ylabel('Momento relativo en Y ($p_y / p_0$)', fontsize=10)

    # --- PANEL 2: CURVA DE TRANSFERENCIA ENERGÉTICA T(θ) ---
    theta_range = np.linspace(0, 180, 500)
    theta_range_rad = np.radians(theta_range)
    Ep_range = E0_keV / (1.0 + alpha * (1.0 - np.cos(theta_range_rad)))
    T_range = E0_keV - Ep_range

    # Curva teórica global
    ax2.plot(theta_range, T_range, color='#2c3e50', linewidth=2.5, label=r'Energía Transferida $T(\theta)$')

    # Marcador de estado dinámico
    ax2.plot(theta_deg, T, 'ro', markersize=9, zorder=5,
             label=f'Ángulo actual: {theta_deg}°\nTransferencia: {T:.1f} keV ({T/E0_keV*100:.1f}%)')

    # Línea horizontal indicando el Borde Compton
    ax2.axhline(T_max, color='#c44e52', linestyle=':', linewidth=1.8,
               label=f'Borde Compton ($T_{{max}} = {T_max:.1f}$ keV)')

    # Guías visuales punteadas para el marcador de estado
    ax2.axvline(theta_deg, color='gray', linestyle='--', alpha=0.5, zorder=2)
    ax2.axhline(T, color='gray', linestyle='--', alpha=0.5, zorder=2)

    # Ajustes del panel de la curva cartesiana
    ax2.set_xlim(0, 180)
    ax2.set_ylim(0, E0_keV * 1.05)
    ax2.set_xlabel('Ángulo de Dispersión del Fotón θ (grados)', fontsize=11)
    ax2.set_ylabel('Energía Cinética del Electrón T (keV)', fontsize=11)
    ax2.set_title('Energía Transferida al Electrón vs Ángulo de Dispersión', fontsize=12, fontweight='bold', pad=10)
    ax2.legend(loc='upper left', fontsize=10)

    if use_sns:
        sns.despine(fig=fig)
    plt.tight_layout()
    plt.show()

# 3. CONSTRUCCIÓN DE LA INTERFAZ DE USUARIO CON IPYWIDGETS
E0_slider = widgets.FloatSlider(
    value=600.2, min=100.0, max=2000.0, step=10.0,
    description='E0 (keV):', continuous_update=True,
    layout=widgets.Layout(width='60%')
)

theta_slider = widgets.IntSlider(
    value=120, min=0, max=180, step=1,
    description='Ángulo θ (°):', continuous_update=True,
    layout=widgets.Layout(width='60%')
)

interactive_widget = widgets.interactive_output(
    simular_compton,
    {'E0_keV': E0_slider, 'theta_deg': theta_slider}
)

# Renderizado de la caja de controles y el widget interactivo
controls = widgets.VBox([E0_slider, theta_slider])
display(controls, interactive_widget)
```
