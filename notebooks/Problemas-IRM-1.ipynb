### **Problema 1**
Supongamos un haz de fotones de 1 MeV incidiendo sobre un disco de aluminio ($\rho$ = 2.70 g cm$^{-3}$) de 10 cm de espesor y 4 cm de diámetro. La sección eficaz total de interacción es $\sigma$=2.75$\times$10$^{-24}$cm$^{2}$ y el peso atómico Pa(Al)=27.0 g.

a) Si un solo fotón incide perpendicularmente sobre la lámina ¿Qué probabilidad tiene de interaccionar dentro del primer centímetro de recorrido? ¿Qué probabilidad tiene de interaccionar antes de escapar del blanco?.

b) Un haz cilíndrico de 1.2$\times$10$^4$ fotones cm$^{-2}$ y 2 cm de diámetro incide perpendicularmente en el centro del disco de aluminio ¿Cuantos fotones del haz interaccionan con el blanco? Resolver el mismo problema para un haz de 10 cm de diámetro (mayor que el blanco).




---


#### a) Probabilidad de interacción de un solo fotón

Para analizar la probabilidad de interacción de un único fotón en el disco de aluminio, primero calculamos el coeficiente de atenuación lineal ($\mu$). Utilizando el número de Avogadro proporcionado en los datos generales de la hoja ($N_a = 6.02 \times 10^{23}$), la densidad atómica es $n = \frac{N_a \cdot \rho}{P_a} = \frac{(6.02 \times 10^{23}) \cdot 2.70}{27.0} = 6.02 \times 10^{22} \text{ átomos/cm}^3$. Multiplicando por la sección eficaz total ($\sigma = 2.75 \times 10^{-24} \text{ cm}^2$), obtenemos un coeficiente **$\mu = 0.16555 \text{ cm}^{-1}$**.

Al evaluar las probabilidades solicitadas para este haz de 1 MeV, podemos comparar el uso de la aproximación matemática de **blanco fino** frente al cálculo exacto de **blanco grueso**:

### 1. Aproximación de blanco fino ($P \approx \mu \cdot x$)
Esta aproximación se deriva del desarrollo en serie de Taylor de la exponencial y asume que la probabilidad de interacción crece linealmente con el espesor. Matemáticamente solo es válida cuando $\mu x \ll 1$ (es decir, cuando el haz apenas sufre atenuación).

*   **Probabilidad en el primer centímetro ($x = 1 \text{ cm}$):**
    $P \approx 0.16555 \cdot 1 = \mathbf{0.1655 \text{ (16.55\%)}}$
*   **Probabilidad antes de escapar del blanco ($x = 10 \text{ cm}$):**
    $P \approx 0.16555 \cdot 10 = \mathbf{1.6555 \text{ (165.55\%)}}$

Como se puede observar, para 10 cm de espesor el resultado de esta aproximación pierde todo el sentido físico (arroja una probabilidad mayor al 100%). El material es demasiado masivo para sostener la aproximación lineal.

### 2. Cálculo exacto o de blanco grueso ($P = 1 - e^{-\mu x}$)
Este cálculo aplica rigurosamente la ley de atenuación exponencial, la cual contabiliza la reducción exponencial de los fotones del haz a medida que van penetrando en el medio. Es válido para cualquier espesor.

*   **Probabilidad en el primer centímetro ($x = 1 \text{ cm}$):**
    $P = 1 - e^{-0.16555\text{ cm}^{-1} \cdot 10 \text{ cm}} = \mathbf{0.1526 \text{ (15.26\%)}}$
*   **Probabilidad antes de escapar del blanco ($x = 10 \text{ cm}$):**
    $P = 1 - e^{-1.6555 \text{ cm}^{-1} \cdot 10 \text{ cm}} = 1 - 0.1910 = \mathbf{0.8090 \text{ (80.90\%)}}$.

### Conclusión de la comparación
Para el **primer centímetro**, la aproximación de blanco fino (16.55%) ofrece una estimación moderadamente cercana al valor real (15.26%), aunque lo sobrestima por no tener en cuenta a los fotones que ya han interaccionado en las capas más superficiales de ese primer centímetro.

Sin embargo, al evaluar todo el **blanco de 10 cm de espesor**, el régimen de "blanco fino" fracasa por completo. Para calcular interacciones en un espesor donde $\mu x$ no es despreciable (en este caso $\mu x = 1.6555$), **es estrictamente necesario aplicar el tratamiento de blanco grueso**, obteniendo así la verdadera probabilidad de interacción del **80.90%**.


```python
# Coeficiente de atenuación lineal calculado (Aluminio a 1 MeV)
mu = 0.16555  # cm^-1

# Rango de profundidades x en centímetros (de 0 a 10 cm)
x = np.linspace(0, 10, 100)

# 1. Aproximación de blanco fino (Probabilidad lineal)
# Nota: Carece de sentido físico cuando supera 1 (100%)
P_fino = mu * x

# 2. Cálculo exacto de blanco grueso (Probabilidad exponencial)
P_grueso = 1 - np.exp(-mu * x)

# Configuración y creación del gráfico
plt.figure(figsize=(9, 6))

# Dibujar las curvas
plt.plot(x, P_fino, label=r'Blanco Fino ($P \approx \mu \cdot x$)',
         linestyle='--', color='red', linewidth=2)
plt.plot(x, P_grueso, label=r'Blanco Grueso ($P = 1 - e^{-\mu \cdot x}$)',
         linestyle='-', color='blue', linewidth=2)

# Línea horizontal que marca el límite físico lógico (100% de probabilidad)
plt.axhline(1, color='gray', linestyle=':', label='Límite físico (Probabilidad = 1)')

# Etiquetas, título y leyenda
plt.title('Comparación de Probabilidad de Interacción: Blanco Fino vs. Grueso\n' + r'(Fotones de 1 MeV en Aluminio, $\mu = 0.16555$ cm$^{-1}$)')
plt.xlabel('Profundidad $x$ (cm)')
plt.ylabel('Probabilidad de Interacción $P(x)$')

# Limitar los ejes para una mejor visualización de la divergencia
plt.xlim(0, 10)
plt.ylim(0, 1.8)

plt.legend(loc='upper left')
plt.grid(True, alpha=0.5)

# Mostrar la gráfica
plt.tight_layout()
plt.show()
```

---

#### b) Fotones del haz que interaccionan con el blanco

Tenemos una fluencia incidente $\Phi = 1.2 \times 10^4 \text{ fotones/cm}^2$. El número de fotones que interaccionan dependerá de cuántos fotones *impacten efectivamente* en el blanco, multiplicado por la probabilidad de interacción calculada antes para los 10 cm de espesor ($P = 0.8090$).

**Caso 1: Haz de 2 cm de diámetro**

Como el diámetro del haz (2 cm) es **menor** que el diámetro del blanco (4 cm), todos los fotones del haz inciden en el disco de aluminio.
*   **Área transversal del haz:** $A_{haz} = \pi \cdot r^2 = \pi \cdot (1 \text{ cm})^2 = \pi \text{ cm}^2 \approx 3.1416 \text{ cm}^2$
*   **Fotones totales que inciden:** $N_{inc} = \Phi \cdot A_{haz} = (1.2 \times 10^4) \cdot \pi \approx 37,699 \text{ fotones}$
*   **Fotones que interaccionan:** $N_{int} = N_{inc} \cdot P(10\text{ cm}) = 37,699 \cdot 0.8090 \approx \mathbf{30,499 \text{ fotones}}$

**Caso 2: Haz de 10 cm de diámetro**

En este caso, el haz (10 cm) es **mayor** que el disco de aluminio (4 cm). Los fotones que caen por fuera del disco no pueden interaccionar. Por lo tanto, el área efectiva de incidencia viene limitada por el propio blanco.
*   **Área efectiva de incidencia (el blanco):** $A_{blanco} = \pi \cdot r_{blanco}^2 = \pi \cdot (2 \text{ cm})^2 = 4\pi \text{ cm}^2 \approx 12.566 \text{ cm}^2$
*   **Fotones que logran golpear el blanco:** $N_{inc} = \Phi \cdot A_{blanco} = (1.2 \times 10^4) \cdot 4\pi \approx 150,796 \text{ fotones}$
*   **Fotones que interaccionan:** $N_{int} = N_{inc} \cdot P(10\text{ cm}) = 150,796 \cdot 0.8090 \approx \mathbf{121,995 \text{ fotones}}$

```python
# Constantes del problema 1 (Aluminio a 1 MeV)
mu = 0.16555  # cm^-1 (coeficiente de atenuación)
phi = 1.2e4   # fotones/cm^2 (fluencia del haz)

def draw_cylinder_3d(ax, radius, z_min, z_max, color, alpha, label=''):
    """Función auxiliar para generar un cilindro en 3D"""
    z = np.linspace(z_min, z_max, 50)
    theta = np.linspace(0, 2*np.pi, 50)
    theta_grid, z_grid = np.meshgrid(theta, z)
    x_grid = radius * np.cos(theta_grid)
    y_grid = radius * np.sin(theta_grid)
    ax.plot_surface(x_grid, y_grid, z_grid, color=color, alpha=alpha, edgecolor='none', label=label)

def calcular_interacciones_3d(d_haz, d_blanco, x_espesor):
    # 1. Cálculos físicos exactos (blanco grueso)
    P = 1 - np.exp(-mu * x_espesor)

    # Lógica del diámetro efectivo
    d_efectivo = min(d_haz, d_blanco)
    area_efectiva = np.pi * (d_efectivo / 2)**2
    n_inc = phi * area_efectiva
    n_int = P * n_inc

    # 2. Mostrar resultados numéricos por pantalla
    print("-" * 60)
    print(f"Probabilidad de interacción (espesor {x_espesor} cm): {P*100:.2f}%")

    if d_haz > d_blanco:
        print(f"\n¡Atención! El haz ({d_haz} cm) desborda el blanco ({d_blanco} cm).")
        print(f"El volumen de interacción queda limitado por el blanco.")
    else:
        print(f"\nEl haz penetra completamente dentro del blanco.")

    print(f"\nFotones incidentes sobre el blanco: {n_inc:,.0f}")
    print(f"Fotones que interaccionan:          {n_int:,.0f}")
    print("-" * 60)

    # 3. Representación Gráfica 3D
    fig = plt.figure(figsize=(10, 8))
    ax = fig.add_subplot(111, projection='3d')

    # A) Dibujar el blanco (Aluminio) - Cilindro Gris
    draw_cylinder_3d(ax, radius=d_blanco/2, z_min=0, z_max=x_espesor, color='gray', alpha=0.3)

    # B) Dibujar el haz incidente - Cilindro Azul (más largo para ver cómo entra/sale)
    draw_cylinder_3d(ax, radius=d_haz/2, z_min=-3, z_max=x_espesor+3, color='blue', alpha=0.15)

    # C) Dibujar el volumen efectivo de interacción - Cilindro Rojo intenso
    draw_cylinder_3d(ax, radius=d_efectivo/2, z_min=0.01, z_max=x_espesor-0.01, color='red', alpha=0.6)

    # Ajustes de los ejes para mantener la proporción geométrica
    limite_xy = max(d_haz, d_blanco) / 2 + 1
    ax.set_xlim(-limite_xy, limite_xy)
    ax.set_ylim(-limite_xy, limite_xy)
    ax.set_zlim(-3, x_espesor + 3)

    # Etiquetas de los ejes
    ax.set_xlabel('Eje X (cm)')
    ax.set_ylabel('Eje Y (cm)')
    ax.set_zlabel('Profundidad Z (cm)')
    ax.set_title('Visualización 3D: Interacción Haz de Fotones vs Blanco', pad=20)

    # Ángulo de cámara inicial (Elevación y Azimut)
    ax.view_init(elev=20, azim=40)

    plt.show()

# Lanzar los controles interactivos
widgets.interact(calcular_interacciones_3d,
                 d_haz=widgets.FloatSlider(value=2.0, min=1.0, max=15.0, step=0.5, description='d_haz (cm):'),
                 d_blanco=widgets.FloatSlider(value=4.0, min=1.0, max=15.0, step=0.5, description='d_blanco (cm):'),
                 x_espesor=widgets.FloatSlider(value=10.0, min=0.5, max=20.0, step=0.5, description='Espesor x (cm):'))
```

### **Problema 2**

Para fotones de 150 keV incidiendo sobre Cu (Z=29, Pa=63.54), la sección eficaz de scattering coherente es de 2.20 $\times$ 10$^{-24}$ cm$^{2}$ y la sección eficaz fotoeléctrica es 8.89 $\times$ 10$^{-24}$ cm$^{2}$. Se sabe además que para esa energía las secciones eficaces Compton (colisión fotón - electrón) de scattering y de transferencia valen 0.363 $\times$ 10$^{-24}$ cm$^{2}$ y 0.081 $\times$ 10$^{-24}$ cm$^{2}$ respectivamente.

Suponiendo un fotón individual que incide sobre una lámina de cobre ($\rho$ = 8.96 g cm$^{-3}$) de 0.5 cm de espesor, hallar la probabilidad de que:

a) el fotón sea dispersado elásticamente.

b) el fotón sea dispersado inelásticamente cediendo parte de su energía a un electrón del medio.

c) el fotón sea totalmente absorbido.

d) Hallar la energía media transferida por colisión.


---
#### Datos iniciales y preparación de las secciones eficaces**
Tenemos un fotón de $E = 150 \text{ keV}$ incidiendo en una lámina de Cobre (Cu) con $Z=29$, $P_a = 63.54 \text{ g}$ y $\rho = 8.96 \text{ g/cm}^3$, con un espesor de $x = 0.5 \text{ cm}$.

Primero, calculamos las secciones eficaces macroscópicas y totales. El enunciado nos da la sección eficaz del efecto Compton por *electrón* ($\sigma_{sc}$ y $\sigma_{tr}$). Dado que el scattering incoherente ocurre con todos los electrones del átomo, multiplicamos por $Z$:
*   **Sección eficaz incoherente (Compton por átomo):**
    $\sigma_{inc} = Z \times (\sigma_{sc} + \sigma_{tr}) = 29 \times (0.363 + 0.081) \times 10^{-24} \text{ cm}^2 = \mathbf{12.9 \times 10^{-24} \text{ cm}^2}$.
*   **Sección eficaz fotoeléctrica:** $\sigma_{ph} = \mathbf{8.89 \times 10^{-24} \text{ cm}^2}$.
*   **Sección eficaz de scattering coherente:** $\sigma_{coh} = \mathbf{2.20 \times 10^{-24} \text{ cm}^2}$.

**Sección eficaz total ($\sigma$):**
Sumamos las contribuciones de todos los procesos:
$\sigma = \sigma_{coh} + \sigma_{ph} + \sigma_{inc} = (2.20 + 8.89 + 12.9) \times 10^{-24} \text{ cm}^2 = \mathbf{24.0 \times 10^{-24} \text{ cm}^2}$.

#### Cálculo de la Densidad Atómica ($N$) y el exponente de atenuación**
El número de átomos por cm$^3$ en el Cobre se obtiene mediante el número de Avogadro:
$N = \frac{N_A \cdot \rho}{P_a} = \frac{(6.02 \times 10^{23}) \cdot 8.96}{63.54} = \mathbf{8.49 \times 10^{22} \text{ cm}^{-3}}$.

El exponente de la ley de atenuación para el espesor de 0.5 cm será:
$N \cdot \sigma \cdot x = (8.49 \times 10^{22}) \times (24.0 \times 10^{-24}) \times 0.5 = \mathbf{1.0188}$

---

#### Probabilidades de interacción (Apartados a, b y c)
> **⚠️ Nota metodológica:** Un error común es intentar calcular la probabilidad individual de un proceso usando $P_i = 1 - e^{-N \sigma_i x}$ (que daría resultados incorrectos del 9%, 42% y 31% respectivamente). La forma correcta, rigurosa para blanco grueso, es calcular la *probabilidad total* de que haya una interacción y, a continuación, multiplicarla por el *peso relativo* ($\sigma_i / \sigma$) del proceso específico.

*   **Probabilidad Total de interacción:**
    $P_{tot} = 1 - \exp(-N \sigma x) = 1 - e^{-1.0188} = \mathbf{0.639}$ (es decir, el 63.9% de los fotones interaccionan).

A continuación, calculamos la probabilidad de cada proceso específico $P_i = P_{tot} \times \left(\frac{\sigma_i}{\sigma}\right)$:

**a) Dispersión elástica (Scattering Coherente):**

La proporción del proceso es $\frac{2.20}{24.0} \approx 0.09$.
$P_{coh} = 0.639 \times 0.09 = \mathbf{0.06 \text{ (6\%)}}$.

**b) Dispersión inelástica cediendo energía (Scattering Incoherente / Compton):**

La proporción del proceso es $\frac{12.9}{24.0} \approx 0.54$.
$P_{inc} = 0.639 \times 0.54 = \mathbf{0.34 \text{ (34\%)}}$.

**c) Absorción total (Efecto Fotoeléctrico):**

La proporción del proceso es $\frac{8.89}{24.0} \approx 0.37$.
$P_{ph} = 0.639 \times 0.37 = \mathbf{0.24 \text{ (24\%)}}$.

---

**d) Energía media transferida por colisión ($\bar{E}_{tr}$)**

Para calcular la energía media que el fotón transfiere al medio **cuando colisiona**, debemos ponderar la energía que se transfiere en cada tipo de interacción por la probabilidad relativa ($\sigma_i/\sigma$) de que ocurra dicha interacción:

1.  **Scattering Coherente:** El fotón solo cambia de dirección sin ceder energía. $E_{tr}^{coh} = \mathbf{0 \text{ keV}}$.
2.  **Efecto Fotoeléctrico:** El fotón es absorbido por completo. $E_{tr}^{ph} = \mathbf{150 \text{ keV}}$.
3.  **Efecto Compton:** La energía transferida viene dada por la relación entre las secciones de transferencia y scattering:
    $E_{tr}^{inc} = E_0 \times \left( \frac{\sigma_{tr}}{\sigma_{tr} + \sigma_{sc}} \right) = 150 \times \left( \frac{0.081}{0.363 + 0.081} \right) = \mathbf{27.4 \text{ keV}}$.

**Cálculo ponderado:**
$$\bar{E}_{tr} = \sum \left( E_{tr,i} \times \frac{\sigma_i}{\sigma} \right)$$
$$\bar{E}_{tr} = (0 \cdot 0.09) + (150 \cdot 0.37) + (27.4 \cdot 0.54)$$
$$\bar{E}_{tr} = 0 + 55.5 + 13.7 = \mathbf{69.2 \text{ keV}}$$

*(La suma total confirma que, en promedio, un fotón de 150 keV que colisiona transferirá 69.2 keV a los electrones del Cobre).*

```python
# Datos fijos del Problema 2 (Cobre a 150 keV)
N = 8.49e22  # Densidad atómica (cm^-3)
sigma_coh = 2.20e-24  # cm^2
sigma_ph = 8.89e-24   # cm^2
# Calculamos incoherente como Z * (sc + tr)
sigma_inc = 29 * (0.363 + 0.081) * 1e-24  # 12.876e-24 cm^2
sigma_tot = sigma_coh + sigma_ph + sigma_inc  # ~ 23.97e-24 cm^2

def analizar_probabilidades_cu(x_espesor):
    # 1. Fórmula Correcta: Probabilidad total ponderada
    P_tot = 1 - np.exp(-N * sigma_tot * x_espesor)

    P_corr = {
        'Scatt. Coherente': P_tot * (sigma_coh / sigma_tot),
        'Fotoeléctrico': P_tot * (sigma_ph / sigma_tot),
        'Scatt. Incoherente': P_tot * (sigma_inc / sigma_tot)
    }

    # 2. Fórmula Incorrecta (aplicando exponencial a cada proceso por separado)
    P_incorr = {
        'Scatt. Coherente': 1 - np.exp(-N * sigma_coh * x_espesor),
        'Fotoeléctrico': 1 - np.exp(-N * sigma_ph * x_espesor),
        'Scatt. Incoherente': 1 - np.exp(-N * sigma_inc * x_espesor)
    }

    # --- Interfaz de Texto ---
    print(f"--- Análisis para lámina de Cobre de {x_espesor:.2f} cm ---")
    print(f"Probabilidad TOTAL de que haya interacción: {P_tot*100:.1f}%\n")
    print(f"{'Proceso':<20} | {'Fórmula CORRECTA':<18} | {'Fórmula INCORRECTA':<18}")
    print("-" * 62)
    for proc in P_corr:
        print(f"{proc:<20} | {P_corr[proc]*100:>15.1f} % | {P_incorr[proc]*100:>16.1f} %")

    # --- Gráfico de Barras Comparativo ---
    fig, ax = plt.subplots(figsize=(9, 5))

    etiquetas = list(P_corr.keys())
    valores_corr = [P_corr[k]*100 for k in etiquetas]
    valores_incorr = [P_incorr[k]*100 for k in etiquetas]

    x = np.arange(len(etiquetas))
    ancho = 0.35

    # Dibujar barras
    ax.bar(x - ancho/2, valores_corr, ancho, label=r'Correcta: $P_{tot} \cdot (\sigma_i / \sigma)$', color='#2ca02c')
    ax.bar(x + ancho/2, valores_incorr, ancho, label=r'Incorrecta: $1 - e^{-N \sigma_i x}$', color='#d62728', alpha=0.7)

    # Configuración estética
    ax.set_ylabel('Probabilidad (%)')
    ax.set_title(f'Impacto del cálculo en blanco grueso (Espesor = {x_espesor} cm)')
    ax.set_xticks(x)
    ax.set_xticklabels(etiquetas)
    ax.legend(loc='upper left')
    ax.grid(axis='y', linestyle=':', alpha=0.7)

    # Ajuste dinámico del eje Y
    ax.set_ylim(0, max(max(valores_corr), max(valores_incorr)) + 15)

    plt.tight_layout()
    plt.show()

# Lanzar los controles interactivos (valor por defecto 0.5 cm como en el problema)
widgets.interact(analizar_probabilidades_cu,
                 x_espesor=widgets.FloatSlider(value=0.5, min=0.01, max=4.0, step=0.05, description='Espesor x (cm):', continuous_update=False))
```

### **Problema 3**

La sección eficaz total de scattering Compton por electrón libre viene dada por la expresión de Klein-Nishina:

$$ \sigma_{KN} = 2\pi r_0^2 \left[ \frac{1 + \gamma}{\gamma^2} \left( \frac{2(1 + \gamma)}{1 + 2\gamma} - \frac{\ln(1 + 2\gamma)}{\gamma} \right) + \frac{\ln(1 + 2\gamma)}{2\gamma} - \frac{1 + 3\gamma}{(1 + 2\gamma)^2} \right] $$

donde $\gamma = \frac{E}{m_e c^2}$.

**Estudiar la validez del efecto Compton** para describir el scattering incoherente $\sigma_{incoh}$ de fotones en los átomos de Carbono (C) y Plomo (Pb).

Para ello, **representar gráficamente** la diferencia:

$$ \frac{Z \cdot \sigma_{KN} - \sigma_{incoh}}{\langle\sigma\rangle} $$

frente a la energía del fotón, donde

$$ \langle\sigma\rangle = \frac{1}{2} ( Z \cdot \sigma_{KN} + \sigma_{incoh}) $$

¿Qué **proceso de interacción domina** para las energías en las que las secciones eficaces Compton e incoherente son sensiblemente diferentes?

*Nota: Las secciones eficaces atómicas pueden encontrarse en la base de datos de la web del NIST: [http://physics.nist.gov/PhysRefData/Xcom/html/xcom1.html](http://physics.nist.gov/PhysRefData/Xcom/html/xcom1.html).*

Para comprender el alcance y la validez del modelo de Klein-Nishina en el cálculo del scattering incoherente, es necesario analizar los fenómenos físicos subyacentes que diferencian a un electrón libre de uno ligado a un átomo.

#### **1. El Efecto de Ligadura Atómica y la Supresión del Scattering**
La sección eficaz de Klein-Nishina ($\sigma_{KN}$) se deriva bajo la hipótesis de que el fotón interacciona con un **electrón libre y en reposo**. En esta aproximación, la sección eficaz atómica se estima simplemente sumando la contribución de cada uno de los $Z$ electrones de manera independiente:
$$ \sigma_{incoh} \approx Z \cdot \sigma_{KN} $$

Sin embargo, en un átomo real, los electrones están **ligados** en diferentes capas con energías de ligadura $E_b$ específicas.
* **A bajas energías ($E \lesssim E_b$):** El fotón no puede transferir cualquier cantidad de energía al electrón debido a las restricciones de los niveles energéticos cuantizados del átomo. El proceso de scattering incoherente se ve fuertemente **suprimido** (lo que experimentalmente se modela multiplicando por la función de dispersión incoherente $S(q, Z)$, que tiende a cero a bajas energías). Por tanto, la sección eficaz real $\sigma_{inc}$ es mucho menor que la predicción teórica: $\sigma_{inc} \ll Z\sigma_{KN}$.
* **A altas energías ($E \gg E_b$):** La energía del fotón es tan superior a la energía de ligadura de los electrones (incluso de la capa K) que estos pueden considerarse efectivamente "libres y en reposo". En este régimen, la aproximación teórica converge perfectamente con la realidad: $\sigma_{inc} \approx Z\sigma_{KN}$.


#### **2. Análisis por Elemento**

* **Carbono ($Z = 6$, elemento ligero):**
  Al tener electrones débilmente ligados (la energía de ligadura de la capa K es de apenas $0.28\text{ keV}$), la aproximación de electrón libre se vuelve excelente ($<5\%$ de diferencia) a partir de energías muy bajas, aproximadamente **$E > 20\text{ keV}$**.
  Por debajo de esta energía, la diferencia se dispara rápidamente hacia el $200\%$. Sin embargo, en esta región de bajas energías el proceso dominante es, con gran diferencia, el **efecto fotoeléctrico** ($\sigma_{ph} \gg \sigma_{inc}$).

* **Plomo ($Z = 82$, elemento pesado):**
  Al tener electrones muy fuertemente ligados (la energía de ligadura de la capa K es de $88\text{ keV}$), los efectos de ligadura se hacen notar hasta energías mucho más elevadas. La aproximación no empieza a ser buena ($<5\%$ de diferencia) hasta que superamos aproximadamente los **$E > 500\text{ keV}$** (0.5 MeV).
  Al igual que en el Carbono, en la zona donde la aproximación Compton falla (bajas energías), el **efecto fotoeléctrico** es el proceso de interacción absolutamente dominante en el Plomo.

#### **3. Conclusión sobre la Validez de la Aproximación**
A efectos prácticos de física de radiaciones y blindajes, **la aproximación de describir el scattering incoherente mediante $Z \cdot \sigma_{KN}$ es plenamente válida**.

Esto se debe a una coincidencia:
1. En el rango de energías donde la aproximación es **incorrecta** (bajas energías), el error cometido es irrelevante porque el **efecto fotoeléctrico absorbe prácticamente el 100% de los fotones**, haciendo que la contribución del scattering a la atenuación total sea despreciable.
2. En el rango de energías donde el **scattering incoherente es el proceso dominante**, la energía del fotón es tan alta que los electrones se comportan como libres, haciendo que la aproximación teórica de Klein-Nishina sea **extremadamente precisa**.


```python
# ---------------------------------------------------------
# CONSTANTES FÍSICAS
# ---------------------------------------------------------
r0 = 2.8179e-15  # Radio clásico del electrón (m)
barn = 1e-28     # Conversión de m^2 a barns
mec2 = 0.511     # Energía en reposo del electrón (MeV)
NA = 6.0221415e23 # Número de Avogadro

def sigma_KN(E_MeV):
    """
    Fórmula teórica de Klein-Nishina para la sección eficaz total
    de scattering Compton por electrón libre (devuelve barns).
    """
    E_MeV = np.where(E_MeV == 0, 1e-10, E_MeV)
    gamma = E_MeV / mec2

    term1 = ((1 + gamma) / gamma**2) * (2 * (1 + gamma) / (1 + 2*gamma) - np.log(1 + 2*gamma) / gamma)
    term2 = np.log(1 + 2*gamma) / (2 * gamma)
    term3 = (1 + 3*gamma) / (1 + 2*gamma)**2

    sigma_m2 = 2 * np.pi * r0**2 * (term1 + term2 - term3)
    return sigma_m2 / barn

def leer_y_convertir_datos_nist(filepath, A):
    """
    Lee los datos del NIST en cm^2/g y los convierte a barns/átomo
    usando la masa atómica (A) del elemento.
    """
    energies, incoherents, photoelectrics = [], [], []

    # Factor de conversión de cm^2/g a barns/átomo
    # (A / NA) * 10^24 = A * 1.66054
    factor_conversion = A * 1.66054125

    try:
        with open(filepath, 'r') as f:
            for line in f:
                parts = line.strip().split()
                if len(parts) == 8:
                    try:
                        e = float(parts[0])
                        # Extraemos y aplicamos la conversión de unidades
                        inc = float(parts[2]) * factor_conversion
                        ph = float(parts[3]) * factor_conversion

                        energies.append(e)
                        incoherents.append(inc)
                        photoelectrics.append(ph)
                    except ValueError:
                        continue

        return np.array(energies), np.array(incoherents), np.array(photoelectrics)
    except FileNotFoundError:
        print(f"Error: No se encontró el archivo '{filepath}'.")
        return None, None, None

# ---------------------------------------------------------
# FUNCIÓN GENERADORA DE GRÁFICOS
# ---------------------------------------------------------
def analizar_validez_compton_ficheros(archivo_txt, simbolo, Z, A, ylim_sigma):
    """
    Analiza la validez de la aproximación Compton convirtiendo las unidades correctamente.
    """
    E, s_inc, s_ph = leer_y_convertir_datos_nist(archivo_txt, A)

    if E is None:
        return

    Z_s_KN = Z * sigma_KN(E)

    # ====================================================
    # GRÁFICO 1: Secciones Eficaces (barns/átomo) vs Energía
    # ====================================================
    plt.figure(figsize=(8, 6))
    plt.plot(E, Z_s_KN, label=r'Z * $\sigma_{KN}$ (teórica)', color='#4c72b0', linewidth=2.5)
    plt.plot(E, s_inc, label='Incoherente (NIST real)', color='#c44e52', linewidth=2.5, linestyle='--')
    plt.plot(E, s_ph, label='Fotoeléctrico (NIST real)', color='#8ebad9', linewidth=2.5)

    plt.xscale('log')
    plt.yscale('log')
    plt.xlim(1e-3, 1e6)
    plt.ylim(1e-4, ylim_sigma)

    plt.xlabel('Energía (MeV)', fontsize=12)
    plt.ylabel(r'$\sigma$ (barn/átomo)', fontsize=12)
    plt.title(f'{simbolo} - Secciones eficaces', fontsize=14, fontweight='bold')
    plt.grid(True, which="both", ls="-", alpha=0.3)
    plt.legend(loc='upper right', fontsize=11)

    plt.text(0.1, 0.85, simbolo, transform=plt.gca().transAxes,
             fontsize=30, fontweight='bold', bbox=dict(facecolor='white', alpha=0.8, edgecolor='gray'))

    plt.tight_layout()
    plt.show()

    # ====================================================
    # GRÁFICO 2: Diferencia Relativa (%)
    # ====================================================
    diferencia_pct = (Z_s_KN - s_inc) / (0.5 * (Z_s_KN + s_inc)) * 100

    plt.figure(figsize=(6, 6))
    plt.plot(E, diferencia_pct, 'D', color='darkblue', markersize=5,
             label=r'$(Z \cdot \sigma_{KN} - \sigma_{inc}) / <\sigma>$')

    plt.xscale('log')
    plt.xlim(1e-3, 1e1)
    plt.ylim(-5, 200)
    plt.gca().yaxis.set_major_formatter(plt.FuncFormatter(lambda y, _: f'{int(y)}%'))

    plt.xlabel('Energía (MeV)', fontsize=12)
    plt.ylabel('Diferencia Compton - Incoherente', fontsize=12)
    plt.grid(True, which="both", ls="-", alpha=0.3)
    plt.legend()

    plt.text(0.7, 0.7, simbolo, transform=plt.gca().transAxes, fontsize=30, fontweight='bold')

    plt.tight_layout()
    plt.show()

# ---------------------------------------------------------
# EJECUCIÓN
# ---------------------------------------------------------
# Para Carbono (Z=6, A=12.011, límite Y=10^3)
analizar_validez_compton_ficheros('data/C_cross-sections-NIST.txt', simbolo='C', Z=6, A=12.011, ylim_sigma=1e3)

# Para Plomo (Z=82, A=207.2, límite Y=10^4)
analizar_validez_compton_ficheros('data/Pb_cross-sections-NIST.txt', simbolo='Pb', Z=82, A=207.2, ylim_sigma=1e4)
```
